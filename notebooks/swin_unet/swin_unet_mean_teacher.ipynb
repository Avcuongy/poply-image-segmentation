```python
SEED = 42


def set_seed(seed=SEED):
    os.environ["PYTHONHASHSEED"] = str(seed)

    random.seed(seed)
    np.random.seed(seed)

    torch.manual_seed(seed)
    torch.cuda.manual_seed(seed)
    torch.cuda.manual_seed_all(seed)

    torch.backends.cudnn.deterministic = True
    torch.backends.cudnn.benchmark = False

    try:
        torch.use_deterministic_algorithms(True, warn_only=True)
    except Exception:
        pass


def seed_worker(worker_id):
    """Seed Python/NumPy and Albumentations independently in each worker."""
    worker_seed = int(torch.initial_seed() % (2**32))
    random.seed(worker_seed)
    np.random.seed(worker_seed)

    info = get_worker_info()
    if info is None:
        return

    dataset = info.dataset
    transform_names = (
        "transform",
        "shared_geometry",
        "weak_transform",
        "strong_transform",
    )

    for offset, name in enumerate(transform_names):
        transform = getattr(dataset, name, None)
        if transform is not None and hasattr(transform, "set_random_seed"):
            transform.set_random_seed(worker_seed + offset + 1)


set_seed(SEED)
print(f"Global seed: {SEED}")
```

# Data augmentation


```python
def build_labeled_shared_geometry(height=224, width=224, seed=SEED):
    return A.Compose(
        [
            A.Resize(height=height, width=width),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.5),
            A.RandomRotate90(p=0.5),
            A.ShiftScaleRotate(
                shift_limit=(-0.0625, 0.0625),
                scale_limit=(-0.15, 0.15),
                rotate_limit=(-45, 45),
                border_mode=cv2.BORDER_CONSTANT,
                p=0.5,
            ),
            A.OneOf(
                [
                    A.ElasticTransform(
                        alpha=120,
                        sigma=6.0,
                        p=1.0,
                    ),
                    A.GridDistortion(p=1.0),
                    A.OpticalDistortion(
                        distort_limit=(-0.30, 0.30),
                        p=1.0,
                    ),
                ],
                p=0.4,
            ),
        ],
        seed=int(seed),
        strict=True,
    )


def build_unlabeled_shared_geometry(height=224, width=224, seed=SEED):
    return A.Compose(
        [
            A.Resize(height=height, width=width),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.5),
            A.RandomRotate90(p=0.5),
            A.ShiftScaleRotate(
                shift_limit=(-0.0625, 0.0625),
                scale_limit=(-0.10, 0.10),
                rotate_limit=(-20, 20),
                border_mode=cv2.BORDER_CONSTANT,
                p=0.5,
            ),
        ],
        seed=int(seed),
        strict=True,
    )


def build_weak_transform(seed=SEED):
    """Stable Teacher view: shared geometry + normalization only."""
    return A.Compose(
        [
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )


def build_labeled_strong_transform(seed=SEED):
    return A.Compose(
        [
            A.OneOf(
                [
                    A.RandomBrightnessContrast(
                        brightness_limit=(-0.10, 0.10),
                        contrast_limit=(-0.10, 0.10),
                        p=1.0,
                    ),
                    A.HueSaturationValue(
                        hue_shift_limit=(-10, 10),
                        sat_shift_limit=(-15, 15),
                        val_shift_limit=(-10, 10),
                        p=1.0,
                    ),
                ],
                p=0.3,
            ),
            A.OneOf(
                [
                    # Albumentations 2.0.8 uses std_range, not var_limit.
                    A.GaussNoise(
                        std_range=(0.012, 0.022),
                        p=1.0,
                    ),
                    A.GaussianBlur(
                        blur_limit=(3, 5),
                        sigma_limit=(0.5, 1.5),
                        p=1.0,
                    ),
                ],
                p=0.2,
            ),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )


def build_unlabeled_strong_transform(seed=SEED):
    return A.Compose(
        [
            A.OneOf(
                [
                    A.RandomBrightnessContrast(
                        brightness_limit=(-0.20, 0.20),
                        contrast_limit=(-0.20, 0.20),
                        p=1.0,
                    ),
                    A.HueSaturationValue(
                        hue_shift_limit=(-15, 15),
                        sat_shift_limit=(-25, 25),
                        val_shift_limit=(-15, 15),
                        p=1.0,
                    ),
                ],
                p=0.8,
            ),
            A.OneOf(
                [
                    A.GaussNoise(
                        std_range=(0.015, 0.035),
                        p=1.0,
                    ),
                    A.GaussianBlur(
                        blur_limit=(3, 5),
                        sigma_limit=(0.5, 2.0),
                        p=1.0,
                    ),
                ],
                p=0.3,
            ),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )


def build_eval_transform(height=224, width=224, seed=SEED):
    return A.Compose(
        [
            A.Resize(height=height, width=width),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )
```

# Data preparation


```python
CLASS_NAMES = {
    0: "background",
    1: "non-neoplastic",
    2: "neoplastic",
}

CLASS_COLORS = {
    0: (0, 0, 0),
    1: (0, 255, 0),
    2: (255, 0, 0),
}


def decode_ground_truth(mask_rgb):
    mask = np.asarray(mask_rgb)

    if mask.ndim == 2:
        unique_values = np.unique(mask)
        if np.isin(unique_values, [0, 1, 2]).all():
            return mask.astype(np.uint8)
        raise ValueError(
            "Grayscale mask must contain only class IDs 0, 1, 2. "
            f"Found: {unique_values[:20]}"
        )

    if mask.ndim != 3 or mask.shape[-1] != 3:
        raise ValueError(f"Expected HxWx3 RGB mask, got shape={mask.shape}")

    colors = np.array(
        [
            [0, 0, 0],
            [0, 255, 0],
            [255, 0, 0],
        ],
        dtype=np.float32,
    )

    pixels = mask.astype(np.float32)
    diff = pixels[:, :, None, :] - colors[None, None, :, :]
    distances = np.sum(diff**2, axis=-1)
    return np.argmin(distances, axis=-1).astype(np.uint8)


def _pick_column(df, explicit_column=None, candidates=None, required=True):
    if explicit_column is not None:
        if explicit_column not in df.columns:
            raise KeyError(
                f"Column '{explicit_column}' not found. "
                f"Available: {df.columns.tolist()}"
            )
        return explicit_column

    candidates = candidates or []
    normalized = {str(col).strip().lower(): col for col in df.columns}

    for candidate in candidates:
        if candidate.lower() in normalized:
            return normalized[candidate.lower()]

    if len(df.columns) == 1:
        return df.columns[0]

    if required:
        raise KeyError(
            "Cannot infer CSV column automatically. "
            f"Available columns: {df.columns.tolist()}"
        )

    return None


class PolypSegmentationDataset(Dataset):
    IMAGE_COLUMN_CANDIDATES = [
        "image",
        "image_name",
        "image_path",
        "filename",
        "file_name",
        "file",
        "id",
    ]
    MASK_COLUMN_CANDIDATES = [
        "mask",
        "mask_name",
        "mask_path",
        "label",
        "label_path",
        "gt",
        "gt_path",
    ]

    def __init__(
        self,
        dataframe,
        image_dir,
        mask_dir=None,
        transform=None,
        image_col=None,
        mask_col=None,
        has_mask=True,
    ):
        self.df = dataframe.reset_index(drop=True).copy()
        self.image_dir = Path(image_dir)
        self.mask_dir = Path(mask_dir) if mask_dir is not None else None
        self.transform = transform
        self.has_mask = bool(has_mask)

        self.image_col = _pick_column(
            self.df,
            explicit_column=image_col,
            candidates=self.IMAGE_COLUMN_CANDIDATES,
            required=True,
        )

        if self.has_mask:
            if self.mask_dir is None:
                raise ValueError("mask_dir is required when has_mask=True.")

            self.mask_col = _pick_column(
                self.df,
                explicit_column=mask_col,
                candidates=self.MASK_COLUMN_CANDIDATES,
                required=True,
            )

            if self.mask_col == self.image_col:
                raise ValueError(
                    "Mask column resolved to the same column as image column."
                )
        else:
            self.mask_col = None

    @staticmethod
    def _resolve_path(root_dir, value):
        candidate = Path(str(value))
        return candidate if candidate.is_absolute() else Path(root_dir) / str(value)

    def __len__(self):
        return len(self.df)

    def get_image_path(self, index):
        return self._resolve_path(
            self.image_dir,
            self.df.iloc[index][self.image_col],
        )

    def get_mask_path(self, index):
        if not self.has_mask:
            return None
        return self._resolve_path(
            self.mask_dir,
            self.df.iloc[index][self.mask_col],
        )

    def __getitem__(self, index):
        image_path = self.get_image_path(index)
        image_bgr = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
        if image_bgr is None:
            raise FileNotFoundError(f"Cannot read image: {image_path}")

        image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

        if not self.has_mask:
            if self.transform is not None:
                image = self.transform(image=image)["image"]
            return {
                "image": image,
                "image_path": str(image_path),
                "image_name": image_path.name,
            }

        mask_path = self.get_mask_path(index)
        mask_raw = cv2.imread(str(mask_path), cv2.IMREAD_UNCHANGED)
        if mask_raw is None:
            raise FileNotFoundError(f"Cannot read mask: {mask_path}")

        if mask_raw.ndim == 3:
            mask_raw = cv2.cvtColor(mask_raw, cv2.COLOR_BGR2RGB)

        mask = decode_ground_truth(mask_raw)

        if self.transform is not None:
            transformed = self.transform(image=image, mask=mask)
            image = transformed["image"]
            mask = transformed["mask"]

        if not torch.is_tensor(mask):
            mask = torch.from_numpy(mask)

        return {
            "image": image,
            "mask": mask.long(),
            "image_path": str(image_path),
            "mask_path": str(mask_path),
            "image_name": image_path.name,
        }
```

```python
class MeanTeacherLabeledDataset(Dataset):
    IMAGE_COLUMN_CANDIDATES = PolypSegmentationDataset.IMAGE_COLUMN_CANDIDATES
    MASK_COLUMN_CANDIDATES = PolypSegmentationDataset.MASK_COLUMN_CANDIDATES

    def __init__(
        self,
        dataframe,
        image_dir,
        mask_dir,
        shared_geometry,
        weak_transform,
        strong_transform,
        image_col=None,
        mask_col=None,
    ):
        self.df = dataframe.reset_index(drop=True).copy()
        self.image_dir = Path(image_dir)
        self.mask_dir = Path(mask_dir)
        self.shared_geometry = shared_geometry
        self.weak_transform = weak_transform
        self.strong_transform = strong_transform

        self.image_col = _pick_column(
            self.df,
            explicit_column=image_col,
            candidates=self.IMAGE_COLUMN_CANDIDATES,
            required=True,
        )
        self.mask_col = _pick_column(
            self.df,
            explicit_column=mask_col,
            candidates=self.MASK_COLUMN_CANDIDATES,
            required=True,
        )

        if self.image_col == self.mask_col:
            raise ValueError("image_col and mask_col cannot be the same.")

    def __len__(self):
        return len(self.df)

    def get_image_path(self, index):
        value = self.df.iloc[index][self.image_col]
        path = Path(str(value))
        return path if path.is_absolute() else self.image_dir / str(value)

    def get_mask_path(self, index):
        value = self.df.iloc[index][self.mask_col]
        path = Path(str(value))
        return path if path.is_absolute() else self.mask_dir / str(value)

    def __getitem__(self, index):
        image_path = self.get_image_path(index)
        mask_path = self.get_mask_path(index)

        image_bgr = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
        if image_bgr is None:
            raise FileNotFoundError(f"Cannot read image: {image_path}")
        image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

        mask_raw = cv2.imread(str(mask_path), cv2.IMREAD_UNCHANGED)
        if mask_raw is None:
            raise FileNotFoundError(f"Cannot read mask: {mask_path}")
        if mask_raw.ndim == 3:
            mask_raw = cv2.cvtColor(mask_raw, cv2.COLOR_BGR2RGB)
        mask = decode_ground_truth(mask_raw)

        # Geometry is sampled ONCE for image + mask.
        aligned = self.shared_geometry(image=image, mask=mask)
        aligned_image = aligned["image"]
        aligned_mask = aligned["mask"]

        weak_image = self.weak_transform(image=aligned_image)["image"]
        strong_image = self.strong_transform(image=aligned_image)["image"]

        if not torch.is_tensor(aligned_mask):
            aligned_mask = torch.from_numpy(aligned_mask)

        return {
            "weak_image": weak_image,
            "strong_image": strong_image,
            "mask": aligned_mask.long(),
            "image_path": str(image_path),
            "mask_path": str(mask_path),
            "image_name": image_path.name,
        }


class MeanTeacherUnlabeledDataset(Dataset):
    IMAGE_COLUMN_CANDIDATES = PolypSegmentationDataset.IMAGE_COLUMN_CANDIDATES

    def __init__(
        self,
        dataframe,
        image_dir,
        shared_geometry,
        weak_transform,
        strong_transform,
        image_col=None,
    ):
        self.df = dataframe.reset_index(drop=True).copy()
        self.image_dir = Path(image_dir)
        self.shared_geometry = shared_geometry
        self.weak_transform = weak_transform
        self.strong_transform = strong_transform

        self.image_col = _pick_column(
            self.df,
            explicit_column=image_col,
            candidates=self.IMAGE_COLUMN_CANDIDATES,
            required=True,
        )

    def __len__(self):
        return len(self.df)

    def get_image_path(self, index):
        value = self.df.iloc[index][self.image_col]
        path = Path(str(value))
        return path if path.is_absolute() else self.image_dir / str(value)

    def __getitem__(self, index):
        image_path = self.get_image_path(index)
        image_bgr = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
        if image_bgr is None:
            raise FileNotFoundError(f"Cannot read image: {image_path}")

        image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

        # Teacher and Student use exactly the same geometry.
        aligned_image = self.shared_geometry(image=image)["image"]
        weak_image = self.weak_transform(image=aligned_image)["image"]
        strong_image = self.strong_transform(image=aligned_image)["image"]

        return {
            "weak_image": weak_image,
            "strong_image": strong_image,
            "image_path": str(image_path),
            "image_name": image_path.name,
        }


def validate_labeled_dataset(dataset, dataset_name="dataset", max_samples=8):
    if getattr(dataset, "mask_col", None) is None:
        raise ValueError(f"[{dataset_name}] mask_col was not resolved.")
    if dataset.image_col == dataset.mask_col:
        raise ValueError(f"[{dataset_name}] image_col == mask_col.")
    if len(dataset) == 0:
        raise ValueError(f"[{dataset_name}] dataset is empty.")

    sample_count = min(int(max_samples), len(dataset))
    indices = np.unique(np.linspace(0, len(dataset) - 1, num=sample_count, dtype=int))

    seen_classes = set()

    for idx in indices:
        image_path = Path(dataset.get_image_path(int(idx)))
        mask_path = Path(dataset.get_mask_path(int(idx)))

        if not image_path.exists():
            raise FileNotFoundError(f"[{dataset_name}] Missing image: {image_path}")
        if not mask_path.exists():
            raise FileNotFoundError(f"[{dataset_name}] Missing mask: {mask_path}")
        if image_path.resolve() == mask_path.resolve():
            raise ValueError(
                f"[{dataset_name}] Image and mask resolve to the same file: {image_path}"
            )

        mask_raw = cv2.imread(str(mask_path), cv2.IMREAD_UNCHANGED)
        if mask_raw is None:
            raise FileNotFoundError(f"[{dataset_name}] Cannot read mask: {mask_path}")
        if mask_raw.ndim == 3:
            mask_raw = cv2.cvtColor(mask_raw, cv2.COLOR_BGR2RGB)

        decoded = decode_ground_truth(mask_raw)
        unique_classes = set(np.unique(decoded).tolist())

        if not unique_classes.issubset({0, 1, 2}):
            raise ValueError(
                f"[{dataset_name}] Invalid classes {sorted(unique_classes)} "
                f"in {mask_path}"
            )
        seen_classes.update(unique_classes)

    print(
        f"[Mask check] {dataset_name}: {len(indices)} samples OK | "
        f"image_col={dataset.image_col!r} | mask_col={dataset.mask_col!r} | "
        f"classes_seen={sorted(seen_classes)}"
    )


def validate_unlabeled_dataset(dataset, dataset_name="unlabeled", max_samples=8):
    if len(dataset) == 0:
        raise ValueError(f"[{dataset_name}] dataset is empty.")

    sample_count = min(int(max_samples), len(dataset))
    indices = np.unique(np.linspace(0, len(dataset) - 1, num=sample_count, dtype=int))

    for idx in indices:
        image_path = Path(dataset.get_image_path(int(idx)))
        if not image_path.exists():
            raise FileNotFoundError(f"[{dataset_name}] Missing image: {image_path}")

        if cv2.imread(str(image_path), cv2.IMREAD_COLOR) is None:
            raise FileNotFoundError(f"[{dataset_name}] Cannot read image: {image_path}")

    print(
        f"[Image check] {dataset_name}: {len(indices)} samples OK | "
        f"image_col={dataset.image_col!r}"
    )
```

# Architecture


## Swin-Unet


```python
"""
Swin Transformer backbone for Swin-Unet.

Code structure adapted from the official Microsoft Swin Transformer implementation:
https://github.com/microsoft/Swin-Transformer

The default configuration corresponds to Swin-T:
    patch_size = 4
    embed_dim = 96
    depths = (2, 2, 6, 2)
    num_heads = (3, 6, 12, 24)
    window_size = 7

Pretrained checkpoint example:
    swin_tiny_patch4_window7_224.pth

License/source note:
    Microsoft Swin Transformer is released under the MIT License.
"""

from __future__ import annotations

from typing import List, Sequence, Tuple

import torch
import torch.nn as nn
import torch.utils.checkpoint as checkpoint

try:
    from timm.layers import DropPath, to_2tuple, trunc_normal_
except ImportError:
    # Compatibility with older timm versions.
    from timm.models.layers import DropPath, to_2tuple, trunc_normal_


class Mlp(nn.Module):
    def __init__(
        self,
        in_features: int,
        hidden_features: int | None = None,
        out_features: int | None = None,
        act_layer=nn.GELU,
        drop: float = 0.0,
    ):
        super().__init__()
        out_features = out_features or in_features
        hidden_features = hidden_features or in_features

        self.fc1 = nn.Linear(in_features, hidden_features)
        self.act = act_layer()
        self.fc2 = nn.Linear(hidden_features, out_features)
        self.drop = nn.Dropout(drop)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        x = self.fc1(x)
        x = self.act(x)
        x = self.drop(x)
        x = self.fc2(x)
        x = self.drop(x)
        return x


def window_partition(x: torch.Tensor, window_size: int) -> torch.Tensor:
    """
    Args:
        x: [B, H, W, C]

    Returns:
        windows: [num_windows * B, window_size, window_size, C]
    """
    B, H, W, C = x.shape

    if H % window_size != 0 or W % window_size != 0:
        raise ValueError(
            f"Feature resolution ({H}, {W}) must be divisible by "
            f"window_size={window_size}."
        )

    x = x.view(
        B,
        H // window_size,
        window_size,
        W // window_size,
        window_size,
        C,
    )

    windows = (
        x.permute(0, 1, 3, 2, 4, 5).contiguous().view(-1, window_size, window_size, C)
    )
    return windows


def window_reverse(
    windows: torch.Tensor,
    window_size: int,
    H: int,
    W: int,
) -> torch.Tensor:
    """Reverse window_partition."""
    B = int(windows.shape[0] / (H * W / window_size / window_size))

    x = windows.view(
        B,
        H // window_size,
        W // window_size,
        window_size,
        window_size,
        -1,
    )

    x = x.permute(0, 1, 3, 2, 4, 5).contiguous().view(B, H, W, -1)

    return x


class WindowAttention(nn.Module):
    """Window-based multi-head self-attention with relative position bias."""

    def __init__(
        self,
        dim: int,
        window_size: Tuple[int, int],
        num_heads: int,
        qkv_bias: bool = True,
        qk_scale: float | None = None,
        attn_drop: float = 0.0,
        proj_drop: float = 0.0,
    ):
        super().__init__()

        self.dim = dim
        self.window_size = window_size
        self.num_heads = num_heads

        head_dim = dim // num_heads
        self.scale = qk_scale or head_dim**-0.5

        table_size = (2 * window_size[0] - 1) * (2 * window_size[1] - 1)
        self.relative_position_bias_table = nn.Parameter(
            torch.zeros(table_size, num_heads)
        )

        coords_h = torch.arange(window_size[0])
        coords_w = torch.arange(window_size[1])
        coords = torch.stack(torch.meshgrid(coords_h, coords_w, indexing="ij"))
        coords_flatten = torch.flatten(coords, 1)

        relative_coords = coords_flatten[:, :, None] - coords_flatten[:, None, :]
        relative_coords = relative_coords.permute(1, 2, 0).contiguous()
        relative_coords[:, :, 0] += window_size[0] - 1
        relative_coords[:, :, 1] += window_size[1] - 1
        relative_coords[:, :, 0] *= 2 * window_size[1] - 1

        relative_position_index = relative_coords.sum(-1)

        self.register_buffer(
            "relative_position_index",
            relative_position_index,
        )

        self.qkv = nn.Linear(dim, dim * 3, bias=qkv_bias)
        self.attn_drop = nn.Dropout(attn_drop)
        self.proj = nn.Linear(dim, dim)
        self.proj_drop = nn.Dropout(proj_drop)
        self.softmax = nn.Softmax(dim=-1)

        trunc_normal_(self.relative_position_bias_table, std=0.02)

    def forward(
        self,
        x: torch.Tensor,
        mask: torch.Tensor | None = None,
    ) -> torch.Tensor:
        B_, N, C = x.shape

        qkv = (
            self.qkv(x)
            .reshape(
                B_,
                N,
                3,
                self.num_heads,
                C // self.num_heads,
            )
            .permute(2, 0, 3, 1, 4)
        )

        q, k, v = qkv[0], qkv[1], qkv[2]
        q = q * self.scale

        attn = q @ k.transpose(-2, -1)

        relative_position_bias = (
            self.relative_position_bias_table[self.relative_position_index.view(-1)]
            .view(N, N, -1)
            .permute(2, 0, 1)
            .contiguous()
        )

        attn = attn + relative_position_bias.unsqueeze(0)

        if mask is not None:
            nW = mask.shape[0]

            attn = attn.view(
                B_ // nW,
                nW,
                self.num_heads,
                N,
                N,
            )

            attn = attn + mask.unsqueeze(1).unsqueeze(0)

            attn = attn.view(
                -1,
                self.num_heads,
                N,
                N,
            )

        attn = self.softmax(attn)
        attn = self.attn_drop(attn)

        x = (attn @ v).transpose(1, 2).reshape(B_, N, C)

        x = self.proj(x)
        x = self.proj_drop(x)

        return x


class SwinTransformerBlock(nn.Module):
    """Standard W-MSA / SW-MSA Swin Transformer block."""

    def __init__(
        self,
        dim: int,
        input_resolution: Tuple[int, int],
        num_heads: int,
        window_size: int = 7,
        shift_size: int = 0,
        mlp_ratio: float = 4.0,
        qkv_bias: bool = True,
        qk_scale: float | None = None,
        drop: float = 0.0,
        attn_drop: float = 0.0,
        drop_path: float = 0.0,
        act_layer=nn.GELU,
        norm_layer=nn.LayerNorm,
    ):
        super().__init__()

        self.dim = dim
        self.input_resolution = input_resolution
        self.num_heads = num_heads
        self.window_size = window_size
        self.shift_size = shift_size
        self.mlp_ratio = mlp_ratio

        if min(input_resolution) <= window_size:
            self.shift_size = 0
            self.window_size = min(input_resolution)

        if not 0 <= self.shift_size < self.window_size:
            raise ValueError("shift_size must satisfy 0 <= shift_size < window_size")

        self.norm1 = norm_layer(dim)
        self.attn = WindowAttention(
            dim=dim,
            window_size=to_2tuple(self.window_size),
            num_heads=num_heads,
            qkv_bias=qkv_bias,
            qk_scale=qk_scale,
            attn_drop=attn_drop,
            proj_drop=drop,
        )

        self.drop_path = DropPath(drop_path) if drop_path > 0.0 else nn.Identity()

        self.norm2 = norm_layer(dim)

        mlp_hidden_dim = int(dim * mlp_ratio)
        self.mlp = Mlp(
            in_features=dim,
            hidden_features=mlp_hidden_dim,
            act_layer=act_layer,
            drop=drop,
        )

        if self.shift_size > 0:
            H, W = input_resolution

            img_mask = torch.zeros((1, H, W, 1))

            h_slices = (
                slice(0, -self.window_size),
                slice(-self.window_size, -self.shift_size),
                slice(-self.shift_size, None),
            )

            w_slices = (
                slice(0, -self.window_size),
                slice(-self.window_size, -self.shift_size),
                slice(-self.shift_size, None),
            )

            cnt = 0
            for h in h_slices:
                for w in w_slices:
                    img_mask[:, h, w, :] = cnt
                    cnt += 1

            mask_windows = window_partition(
                img_mask,
                self.window_size,
            )

            mask_windows = mask_windows.view(
                -1,
                self.window_size * self.window_size,
            )

            attn_mask = mask_windows.unsqueeze(1) - mask_windows.unsqueeze(2)

            attn_mask = attn_mask.masked_fill(
                attn_mask != 0,
                float(-100.0),
            )
            attn_mask = attn_mask.masked_fill(
                attn_mask == 0,
                float(0.0),
            )
        else:
            attn_mask = None

        self.register_buffer(
            "attn_mask",
            attn_mask,
            persistent=False,
        )

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        H, W = self.input_resolution
        B, L, C = x.shape

        if L != H * W:
            raise ValueError(f"Input token length {L} does not match H*W={H*W}.")

        shortcut = x

        x = self.norm1(x)
        x = x.view(B, H, W, C)

        if self.shift_size > 0:
            shifted_x = torch.roll(
                x,
                shifts=(-self.shift_size, -self.shift_size),
                dims=(1, 2),
            )
        else:
            shifted_x = x

        x_windows = window_partition(
            shifted_x,
            self.window_size,
        )

        x_windows = x_windows.view(
            -1,
            self.window_size * self.window_size,
            C,
        )

        attn_windows = self.attn(
            x_windows,
            mask=self.attn_mask,
        )

        attn_windows = attn_windows.view(
            -1,
            self.window_size,
            self.window_size,
            C,
        )

        shifted_x = window_reverse(
            attn_windows,
            self.window_size,
            H,
            W,
        )

        if self.shift_size > 0:
            x = torch.roll(
                shifted_x,
                shifts=(self.shift_size, self.shift_size),
                dims=(1, 2),
            )
        else:
            x = shifted_x

        x = x.view(B, H * W, C)

        x = shortcut + self.drop_path(x)
        x = x + self.drop_path(self.mlp(self.norm2(x)))

        return x


class PatchMerging(nn.Module):
    """Downsample HxW -> H/2 x W/2 and double channel dimension."""

    def __init__(
        self,
        input_resolution: Tuple[int, int],
        dim: int,
        norm_layer=nn.LayerNorm,
    ):
        super().__init__()

        self.input_resolution = input_resolution
        self.dim = dim

        self.reduction = nn.Linear(
            4 * dim,
            2 * dim,
            bias=False,
        )
        self.norm = norm_layer(4 * dim)

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        H, W = self.input_resolution
        B, L, C = x.shape

        if L != H * W:
            raise ValueError("Input feature has wrong token length.")

        if H % 2 != 0 or W % 2 != 0:
            raise ValueError(f"PatchMerging requires even resolution, got ({H}, {W}).")

        x = x.view(B, H, W, C)

        x0 = x[:, 0::2, 0::2, :]
        x1 = x[:, 1::2, 0::2, :]
        x2 = x[:, 0::2, 1::2, :]
        x3 = x[:, 1::2, 1::2, :]

        x = torch.cat(
            [x0, x1, x2, x3],
            dim=-1,
        )

        x = x.view(B, -1, 4 * C)
        x = self.norm(x)
        x = self.reduction(x)

        return x


class BasicLayer(nn.Module):
    """One hierarchical Swin Transformer stage."""

    def __init__(
        self,
        dim: int,
        input_resolution: Tuple[int, int],
        depth: int,
        num_heads: int,
        window_size: int,
        mlp_ratio: float = 4.0,
        qkv_bias: bool = True,
        qk_scale: float | None = None,
        drop: float = 0.0,
        attn_drop: float = 0.0,
        drop_path: float | Sequence[float] = 0.0,
        norm_layer=nn.LayerNorm,
        downsample=None,
        use_checkpoint: bool = False,
    ):
        super().__init__()

        self.use_checkpoint = use_checkpoint

        if isinstance(drop_path, Sequence):
            drop_path_values = list(drop_path)
        else:
            drop_path_values = [drop_path] * depth

        self.blocks = nn.ModuleList(
            [
                SwinTransformerBlock(
                    dim=dim,
                    input_resolution=input_resolution,
                    num_heads=num_heads,
                    window_size=window_size,
                    shift_size=(0 if i % 2 == 0 else window_size // 2),
                    mlp_ratio=mlp_ratio,
                    qkv_bias=qkv_bias,
                    qk_scale=qk_scale,
                    drop=drop,
                    attn_drop=attn_drop,
                    drop_path=drop_path_values[i],
                    norm_layer=norm_layer,
                )
                for i in range(depth)
            ]
        )

        if downsample is not None:
            self.downsample = downsample(
                input_resolution,
                dim=dim,
                norm_layer=norm_layer,
            )
        else:
            self.downsample = None

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        for block in self.blocks:
            if self.use_checkpoint and self.training:
                x = checkpoint.checkpoint(
                    block,
                    x,
                    use_reentrant=False,
                )
            else:
                x = block(x)

        if self.downsample is not None:
            x = self.downsample(x)

        return x


class PatchEmbed(nn.Module):
    """Image to non-overlapping patch tokens."""

    def __init__(
        self,
        img_size: int | Tuple[int, int] = 224,
        patch_size: int | Tuple[int, int] = 4,
        in_chans: int = 3,
        embed_dim: int = 96,
        norm_layer=None,
    ):
        super().__init__()

        img_size = to_2tuple(img_size)
        patch_size = to_2tuple(patch_size)

        patches_resolution = (
            img_size[0] // patch_size[0],
            img_size[1] // patch_size[1],
        )

        self.img_size = img_size
        self.patch_size = patch_size
        self.patches_resolution = patches_resolution
        self.num_patches = patches_resolution[0] * patches_resolution[1]

        self.proj = nn.Conv2d(
            in_chans,
            embed_dim,
            kernel_size=patch_size,
            stride=patch_size,
        )

        self.norm = norm_layer(embed_dim) if norm_layer is not None else None

    def forward(self, x: torch.Tensor) -> torch.Tensor:
        _, _, H, W = x.shape

        if (H, W) != self.img_size:
            raise ValueError(f"Expected input size {self.img_size}, got ({H}, {W}).")

        x = self.proj(x)
        x = x.flatten(2).transpose(1, 2)

        if self.norm is not None:
            x = self.norm(x)

        return x


class SwinTransformerBackbone(nn.Module):
    """
    Hierarchical Swin Transformer encoder.

    Default parameters = Swin-T ImageNet-1K configuration.
    """

    def __init__(
        self,
        img_size: int = 224,
        patch_size: int = 4,
        in_chans: int = 3,
        embed_dim: int = 96,
        depths: Sequence[int] = (2, 2, 6, 2),
        num_heads: Sequence[int] = (3, 6, 12, 24),
        window_size: int = 7,
        mlp_ratio: float = 4.0,
        qkv_bias: bool = True,
        qk_scale: float | None = None,
        drop_rate: float = 0.0,
        attn_drop_rate: float = 0.0,
        drop_path_rate: float = 0.2,
        norm_layer=nn.LayerNorm,
        ape: bool = False,
        patch_norm: bool = True,
        use_checkpoint: bool = False,
    ):
        super().__init__()

        self.img_size = img_size
        self.patch_size = patch_size
        self.in_chans = in_chans
        self.embed_dim = embed_dim
        self.depths = tuple(depths)
        self.num_heads = tuple(num_heads)
        self.window_size = window_size

        self.num_layers = len(depths)
        self.ape = ape
        self.patch_norm = patch_norm

        self.num_features = int(embed_dim * 2 ** (self.num_layers - 1))

        self.out_channels = tuple(int(embed_dim * 2**i) for i in range(self.num_layers))

        self.patch_embed = PatchEmbed(
            img_size=img_size,
            patch_size=patch_size,
            in_chans=in_chans,
            embed_dim=embed_dim,
            norm_layer=(norm_layer if patch_norm else None),
        )

        num_patches = self.patch_embed.num_patches
        patches_resolution = self.patch_embed.patches_resolution
        self.patches_resolution = patches_resolution

        if ape:
            self.absolute_pos_embed = nn.Parameter(
                torch.zeros(
                    1,
                    num_patches,
                    embed_dim,
                )
            )
            trunc_normal_(
                self.absolute_pos_embed,
                std=0.02,
            )

        self.pos_drop = nn.Dropout(p=drop_rate)

        dpr = [
            x.item()
            for x in torch.linspace(
                0,
                drop_path_rate,
                sum(depths),
            )
        ]

        self.layers = nn.ModuleList()

        for i_layer in range(self.num_layers):
            layer = BasicLayer(
                dim=int(embed_dim * 2**i_layer),
                input_resolution=(
                    patches_resolution[0] // (2**i_layer),
                    patches_resolution[1] // (2**i_layer),
                ),
                depth=depths[i_layer],
                num_heads=num_heads[i_layer],
                window_size=window_size,
                mlp_ratio=mlp_ratio,
                qkv_bias=qkv_bias,
                qk_scale=qk_scale,
                drop=drop_rate,
                attn_drop=attn_drop_rate,
                drop_path=dpr[sum(depths[:i_layer]) : sum(depths[: i_layer + 1])],
                norm_layer=norm_layer,
                downsample=(PatchMerging if i_layer < self.num_layers - 1 else None),
                use_checkpoint=use_checkpoint,
            )

            self.layers.append(layer)

        self.norm = norm_layer(self.num_features)

        self.apply(self._init_weights)

    @staticmethod
    def _init_weights(module: nn.Module) -> None:
        if isinstance(module, nn.Linear):
            trunc_normal_(
                module.weight,
                std=0.02,
            )
            if module.bias is not None:
                nn.init.constant_(
                    module.bias,
                    0,
                )

        elif isinstance(module, nn.LayerNorm):
            nn.init.constant_(
                module.bias,
                0,
            )
            nn.init.constant_(
                module.weight,
                1.0,
            )

    def forward(
        self,
        x: torch.Tensor,
    ) -> Tuple[torch.Tensor, List[torch.Tensor]]:
        """
        Returns:
            bottleneck:
                [B, H/32 * W/32, 768] for Swin-T @ 224.

            skips:
                Token features before each encoder stage:
                stage 0: [B, 56*56, 96]
                stage 1: [B, 28*28, 192]
                stage 2: [B, 14*14, 384]
                stage 3: [B,  7*7,  768]
        """
        x = self.patch_embed(x)

        if self.ape:
            x = x + self.absolute_pos_embed

        x = self.pos_drop(x)

        skips: List[torch.Tensor] = []

        for layer in self.layers:
            skips.append(x)
            x = layer(x)

        x = self.norm(x)

        return x, skips

    def load_pretrained(
        self,
        checkpoint_path: str,
        strict: bool = False,
    ):
        """
        Load an official Microsoft Swin Transformer checkpoint.

        Expected checkpoint:
            swin_tiny_patch4_window7_224.pth

        Classification-only parameters such as `head.*` are ignored.
        The encoder keys keep the same names as Microsoft's implementation.
        """
        checkpoint_data = torch.load(
            checkpoint_path,
            map_location="cpu",
        )

        state_dict = checkpoint_data.get(
            "model",
            checkpoint_data,
        )

        # Remove common wrappers if present.
        cleaned_state_dict = {}

        for key, value in state_dict.items():
            if key.startswith("module."):
                key = key[len("module.") :]

            if key.startswith("backbone."):
                key = key[len("backbone.") :]

            # Classification head is not part of the segmentation backbone.
            if key.startswith("head."):
                continue

            cleaned_state_dict[key] = value

        model_state = self.state_dict()

        compatible_state = {}
        skipped = []

        for key, value in cleaned_state_dict.items():
            if key not in model_state:
                continue

            if model_state[key].shape != value.shape:
                skipped.append(
                    (
                        key,
                        tuple(value.shape),
                        tuple(model_state[key].shape),
                    )
                )
                continue

            compatible_state[key] = value

        msg = self.load_state_dict(
            compatible_state,
            strict=strict,
        )

        print(f"[Swin backbone] Loaded " f"{len(compatible_state)} tensors.")

        if skipped:
            print(
                f"[Swin backbone] Skipped " f"{len(skipped)} shape-mismatched tensors."
            )

        print(f"[Swin backbone] Missing keys: " f"{len(msg.missing_keys)}")
        print(f"[Swin backbone] Unexpected keys: " f"{len(msg.unexpected_keys)}")

        return msg
```

```python
"""
Swin-Unet model using a Swin Transformer encoder backbone.

Architecture adapted from the official Swin-Unet implementation:
https://github.com/HuCaoFighting/Swin-Unet

Encoder / pretrained weights:
https://github.com/microsoft/Swin-Transformer

The default model is Swin-Unet with a Swin-T encoder:
    img_size = 224
    patch_size = 4
    embed_dim = 96
    depths = (2, 2, 6, 2)
    num_heads = (3, 6, 12, 24)
    window_size = 7
"""

from __future__ import annotations

import copy
from typing import Sequence, Tuple

import torch
import torch.nn as nn
import torch.utils.checkpoint as checkpoint


class PatchExpand(nn.Module):
    """2x token-space upsampling used by the Swin-Unet decoder."""

    def __init__(
        self,
        input_resolution: Tuple[int, int],
        dim: int,
        dim_scale: int = 2,
        norm_layer=nn.LayerNorm,
    ):
        super().__init__()

        self.input_resolution = input_resolution
        self.dim = dim
        self.dim_scale = dim_scale

        self.expand = (
            nn.Linear(
                dim,
                2 * dim,
                bias=False,
            )
            if dim_scale == 2
            else nn.Identity()
        )

        self.norm = norm_layer(dim // dim_scale)

    def forward(
        self,
        x: torch.Tensor,
    ) -> torch.Tensor:
        H, W = self.input_resolution

        x = self.expand(x)

        B, L, C = x.shape

        if L != H * W:
            raise ValueError(f"Input token length {L} != H*W={H*W}.")

        x = x.view(B, H, W, C)

        # Equivalent to pixel shuffle in token/channel layout.
        x = x.view(
            B,
            H,
            W,
            2,
            2,
            C // 4,
        )

        x = (
            x.permute(
                0,
                1,
                3,
                2,
                4,
                5,
            )
            .contiguous()
            .view(
                B,
                H * 2,
                W * 2,
                C // 4,
            )
        )

        x = x.view(
            B,
            -1,
            C // 4,
        )

        x = self.norm(x)

        return x


class FinalPatchExpandX4(nn.Module):
    """Final 4x expansion from patch resolution back to image resolution."""

    def __init__(
        self,
        input_resolution: Tuple[int, int],
        dim: int,
        dim_scale: int = 4,
        norm_layer=nn.LayerNorm,
    ):
        super().__init__()

        self.input_resolution = input_resolution
        self.dim = dim
        self.dim_scale = dim_scale

        self.expand = nn.Linear(
            dim,
            (dim_scale**2) * dim,
            bias=False,
        )

        self.norm = norm_layer(dim)

    def forward(
        self,
        x: torch.Tensor,
    ) -> torch.Tensor:
        H, W = self.input_resolution

        x = self.expand(x)

        B, L, C = x.shape

        if L != H * W:
            raise ValueError(f"Input token length {L} != H*W={H*W}.")

        scale = self.dim_scale

        x = x.view(
            B,
            H,
            W,
            scale,
            scale,
            C // (scale**2),
        )

        x = (
            x.permute(
                0,
                1,
                3,
                2,
                4,
                5,
            )
            .contiguous()
            .view(
                B,
                H * scale,
                W * scale,
                C // (scale**2),
            )
        )

        x = x.view(
            B,
            -1,
            self.dim,
        )

        x = self.norm(x)

        return x


class BasicLayerUp(nn.Module):
    """One Swin Transformer decoder stage followed optionally by PatchExpand."""

    def __init__(
        self,
        dim: int,
        input_resolution: Tuple[int, int],
        depth: int,
        num_heads: int,
        window_size: int,
        mlp_ratio: float = 4.0,
        qkv_bias: bool = True,
        qk_scale: float | None = None,
        drop: float = 0.0,
        attn_drop: float = 0.0,
        drop_path: float | Sequence[float] = 0.0,
        norm_layer=nn.LayerNorm,
        upsample=None,
        use_checkpoint: bool = False,
    ):
        super().__init__()

        self.use_checkpoint = use_checkpoint

        if isinstance(drop_path, Sequence):
            drop_path_values = list(drop_path)
        else:
            drop_path_values = [drop_path] * depth

        self.blocks = nn.ModuleList(
            [
                SwinTransformerBlock(
                    dim=dim,
                    input_resolution=input_resolution,
                    num_heads=num_heads,
                    window_size=window_size,
                    shift_size=(0 if i % 2 == 0 else window_size // 2),
                    mlp_ratio=mlp_ratio,
                    qkv_bias=qkv_bias,
                    qk_scale=qk_scale,
                    drop=drop,
                    attn_drop=attn_drop,
                    drop_path=drop_path_values[i],
                    norm_layer=norm_layer,
                )
                for i in range(depth)
            ]
        )

        if upsample is not None:
            self.upsample = upsample(
                input_resolution=input_resolution,
                dim=dim,
                dim_scale=2,
                norm_layer=norm_layer,
            )
        else:
            self.upsample = None

    def forward(
        self,
        x: torch.Tensor,
    ) -> torch.Tensor:
        for block in self.blocks:
            if self.use_checkpoint and self.training:
                x = checkpoint.checkpoint(
                    block,
                    x,
                    use_reentrant=False,
                )
            else:
                x = block(x)

        if self.upsample is not None:
            x = self.upsample(x)

        return x


class SwinUNet(nn.Module):
    """
    Swin-Unet with a configurable Swin encoder.

    Defaults match Swin-T + 3-class semantic segmentation.

    Args:
        img_size:
            Fixed model input size. Official Swin-Unet commonly uses 224.

        num_classes:
            Segmentation classes. Use 3 for:
            background, non-neoplastic, neoplastic.

        pretrained_path:
            Optional Microsoft Swin-T ImageNet-1K checkpoint path.

        pretrained_decoder:
            If True, encoder stage weights are mirrored into compatible
            decoder Swin blocks following the initialization idea used by
            the official Swin-Unet implementation.
    """

    def __init__(
        self,
        img_size: int = 224,
        patch_size: int = 4,
        in_channels: int = 3,
        num_classes: int = 3,
        embed_dim: int = 96,
        depths: Sequence[int] = (2, 2, 6, 2),
        num_heads: Sequence[int] = (3, 6, 12, 24),
        window_size: int = 7,
        mlp_ratio: float = 4.0,
        qkv_bias: bool = True,
        qk_scale: float | None = None,
        drop_rate: float = 0.0,
        attn_drop_rate: float = 0.0,
        drop_path_rate: float = 0.2,
        norm_layer=nn.LayerNorm,
        ape: bool = False,
        patch_norm: bool = True,
        use_checkpoint: bool = False,
        pretrained_path: str | None = None,
        pretrained_decoder: bool = True,
    ):
        super().__init__()

        self.img_size = img_size
        self.patch_size = patch_size
        self.in_channels = in_channels
        self.num_classes = num_classes
        self.embed_dim = embed_dim
        self.depths = tuple(depths)
        self.num_heads = tuple(num_heads)
        self.window_size = window_size
        self.num_layers = len(depths)

        # Encoder: Microsoft Swin Transformer-compatible backbone
        self.encoder = SwinTransformerBackbone(
            img_size=img_size,
            patch_size=patch_size,
            in_chans=in_channels,
            embed_dim=embed_dim,
            depths=depths,
            num_heads=num_heads,
            window_size=window_size,
            mlp_ratio=mlp_ratio,
            qkv_bias=qkv_bias,
            qk_scale=qk_scale,
            drop_rate=drop_rate,
            attn_drop_rate=attn_drop_rate,
            drop_path_rate=drop_path_rate,
            norm_layer=norm_layer,
            ape=ape,
            patch_norm=patch_norm,
            use_checkpoint=use_checkpoint,
        )

        patches_resolution = self.encoder.patches_resolution

        # Same stochastic-depth schedule as encoder.
        dpr = [
            x.item()
            for x in torch.linspace(
                0,
                drop_path_rate,
                sum(depths),
            )
        ]

        # Decoder
        self.layers_up = nn.ModuleList()
        self.concat_back_dim = nn.ModuleList()

        for i_layer in range(self.num_layers):
            encoder_stage = self.num_layers - 1 - i_layer

            dim = int(embed_dim * 2**encoder_stage)

            resolution = (
                patches_resolution[0] // (2**encoder_stage),
                patches_resolution[1] // (2**encoder_stage),
            )

            if i_layer == 0:
                self.concat_back_dim.append(nn.Identity())

                self.layers_up.append(
                    PatchExpand(
                        input_resolution=resolution,
                        dim=dim,
                        dim_scale=2,
                        norm_layer=norm_layer,
                    )
                )

            else:
                self.concat_back_dim.append(
                    nn.Linear(
                        2 * dim,
                        dim,
                    )
                )

                stage_dpr = dpr[
                    sum(depths[:encoder_stage]) : sum(depths[: encoder_stage + 1])
                ]

                self.layers_up.append(
                    BasicLayerUp(
                        dim=dim,
                        input_resolution=resolution,
                        depth=depths[encoder_stage],
                        num_heads=num_heads[encoder_stage],
                        window_size=window_size,
                        mlp_ratio=mlp_ratio,
                        qkv_bias=qkv_bias,
                        qk_scale=qk_scale,
                        drop=drop_rate,
                        attn_drop=attn_drop_rate,
                        drop_path=stage_dpr,
                        norm_layer=norm_layer,
                        upsample=(
                            PatchExpand if i_layer < self.num_layers - 1 else None
                        ),
                        use_checkpoint=use_checkpoint,
                    )
                )

        self.norm_up = norm_layer(embed_dim)

        self.final_up = FinalPatchExpandX4(
            input_resolution=patches_resolution,
            dim=embed_dim,
            dim_scale=4,
            norm_layer=norm_layer,
        )

        self.segmentation_head = nn.Conv2d(
            in_channels=embed_dim,
            out_channels=num_classes,
            kernel_size=1,
            bias=False,
        )

        self.apply(self._init_decoder_weights)

        if pretrained_path is not None:
            self.load_pretrained(
                pretrained_path,
                load_decoder=pretrained_decoder,
            )

    @staticmethod
    def _init_decoder_weights(
        module: nn.Module,
    ) -> None:
        # Encoder initializes itself. Reinitializing it with the same rule is harmless before loading pretrained weights.
        try:
            from timm.layers import trunc_normal_
        except ImportError:
            from timm.models.layers import trunc_normal_

        if isinstance(module, nn.Linear):
            trunc_normal_(
                module.weight,
                std=0.02,
            )
            if module.bias is not None:
                nn.init.constant_(
                    module.bias,
                    0,
                )

        elif isinstance(module, nn.LayerNorm):
            nn.init.constant_(
                module.bias,
                0,
            )
            nn.init.constant_(
                module.weight,
                1.0,
            )

    def forward_up_features(
        self,
        x: torch.Tensor,
        skips: list[torch.Tensor],
    ) -> torch.Tensor:
        for i, layer_up in enumerate(self.layers_up):
            if i == 0:
                x = layer_up(x)
                continue

            skip_index = self.num_layers - 1 - i

            skip = skips[skip_index]

            if x.shape[1] != skip.shape[1]:
                raise ValueError(
                    f"Decoder/skip token mismatch at stage {i}: "
                    f"{x.shape} vs {skip.shape}"
                )

            x = torch.cat(
                [x, skip],
                dim=-1,
            )

            x = self.concat_back_dim[i](x)
            x = layer_up(x)

        x = self.norm_up(x)

        return x

    def tokens_to_feature_map(
        self,
        x: torch.Tensor,
        resolution: Tuple[int, int],
    ) -> torch.Tensor:
        B, L, C = x.shape
        H, W = resolution

        if L != H * W:
            raise ValueError(f"Token length {L} != H*W={H*W}.")

        return x.transpose(1, 2).contiguous().view(B, C, H, W)

    def forward(
        self,
        x: torch.Tensor,
        return_features: bool = False,
    ):
        if x.shape[1] != self.in_channels:
            raise ValueError(
                f"Expected {self.in_channels} input channels, " f"got {x.shape[1]}."
            )

        bottleneck, skips = self.encoder(x)

        decoder_tokens = self.forward_up_features(
            bottleneck,
            skips,
        )

        x = self.final_up(decoder_tokens)

        B, L, C = x.shape

        if L != self.img_size * self.img_size:
            raise ValueError(
                "Final token count does not match output image resolution."
            )

        x = x.view(
            B,
            self.img_size,
            self.img_size,
            C,
        )

        x = x.permute(
            0,
            3,
            1,
            2,
        ).contiguous()

        logits = self.segmentation_head(x)

        if not return_features:
            return logits

        bottleneck_resolution = (
            self.encoder.patches_resolution[0] // (2 ** (self.num_layers - 1)),
            self.encoder.patches_resolution[1] // (2 ** (self.num_layers - 1)),
        )

        bottleneck_map = self.tokens_to_feature_map(
            bottleneck,
            bottleneck_resolution,
        )

        decoder_map = self.tokens_to_feature_map(
            decoder_tokens,
            self.encoder.patches_resolution,
        )

        return logits, {
            "bottleneck": bottleneck_map,
            "decoder": decoder_map,
        }

    def load_pretrained(
        self,
        checkpoint_path: str,
        load_decoder: bool = True,
    ):
        """
        Load Microsoft Swin checkpoint into encoder.

        If load_decoder=True, compatible encoder Swin block weights are
        mirrored into decoder stages, following the official Swin-Unet
        pretraining strategy.
        """

        # 1) Load exact Microsoft-compatible keys into the encoder.
        encoder_msg = self.encoder.load_pretrained(
            checkpoint_path,
            strict=False,
        )

        if not load_decoder:
            return encoder_msg

        # 2) Read checkpoint again to initialize compatible decoder blocks.
        checkpoint_data = torch.load(
            checkpoint_path,
            map_location="cpu",
        )

        state_dict = checkpoint_data.get(
            "model",
            checkpoint_data,
        )

        cleaned = {}

        for key, value in state_dict.items():
            if key.startswith("module."):
                key = key[len("module.") :]

            if key.startswith("backbone."):
                key = key[len("backbone.") :]

            if key.startswith("head."):
                continue

            cleaned[key] = value

        model_state = self.state_dict()

        decoder_state = {}

        # Map encoder layer N -> decoder layer (3-N).
        # Only keys with exact matching names/shapes are loaded.
        for key, value in cleaned.items():
            if not key.startswith("layers."):
                continue

            parts = key.split(".")

            if len(parts) < 3:
                continue

            encoder_stage = int(parts[1])
            decoder_stage = self.num_layers - 1 - encoder_stage

            new_key = f"layers_up.{decoder_stage}." + ".".join(parts[2:])

            if new_key not in model_state:
                continue

            if model_state[new_key].shape != value.shape:
                continue

            decoder_state[new_key] = value

        decoder_msg = self.load_state_dict(
            decoder_state,
            strict=False,
        )

        print(f"[Swin-Unet decoder] Loaded " f"{len(decoder_state)} mirrored tensors.")

        return {
            "encoder": encoder_msg,
            "decoder": decoder_msg,
        }
```

# Metrics and losses


## metrics


```python
def multiclass_dice_score(
    logits,
    targets,
    num_classes=3,
    include_background=False,
    eps=1e-7,
):
    """
    Mean per-class Dice over the batch.

    For each class, samples where both GT and prediction are empty are
    excluded from that class's mean. This prevents rare classes from
    receiving artificial Dice=1 scores simply because the class is absent.
    """
    preds = torch.argmax(logits, dim=1)
    start_class = 0 if include_background else 1
    scores = []

    for class_id in range(start_class, num_classes):
        pred_c = preds == class_id
        target_c = targets == class_id

        intersection = (pred_c & target_c).sum(dim=(1, 2)).float()

        denominator = pred_c.sum(dim=(1, 2)).float() + target_c.sum(dim=(1, 2)).float()

        valid = denominator > 0

        if valid.any():
            score = (2.0 * intersection[valid] + eps) / (denominator[valid] + eps)

            scores.append(score.mean())

    if not scores:
        return torch.tensor(
            0.0,
            device=logits.device,
        )

    return torch.stack(scores).mean()


def hd95_binary(pred_mask, true_mask):
    """
    Compute symmetric HD95 when both GT and prediction contain the class.

    If either side is empty, HD95 is undefined because there are not two
    surfaces to compare, so NaN is returned. Dataset-level evaluation keeps
    using np.nanmean, meaning only valid surface-distance cases contribute.
    """
    pred = np.asarray(pred_mask).astype(bool)
    true = np.asarray(true_mask).astype(bool)

    if not pred.any() or not true.any():
        return np.nan

    pred_surface = pred ^ binary_erosion(pred)
    true_surface = true ^ binary_erosion(true)

    if not pred_surface.any():
        pred_surface = pred

    if not true_surface.any():
        true_surface = true

    dist_to_true = distance_transform_edt(~true_surface)
    dist_to_pred = distance_transform_edt(~pred_surface)

    d1 = dist_to_true[pred_surface]
    d2 = dist_to_pred[true_surface]
    distances = np.concatenate([d1, d2])

    return float(np.percentile(distances, 95))


def _safe_divide(numerator, denominator):
    if denominator == 0:
        return np.nan
    return float(numerator / denominator)
```

## losses


```python
class GeneralizedDiceLoss(nn.Module):
    """Generalized Dice Loss (GDL): Automatically balances class weights based on the inverse square of the true area."""

    def __init__(self, num_classes=3, smooth=1e-6, ignore_index=None):
        super().__init__()
        self.num_classes = num_classes
        self.smooth = smooth
        self.ignore_index = ignore_index

    def forward(self, logits, targets):
        probabilities = torch.softmax(logits, dim=1)

        if self.ignore_index is not None:
            valid_mask = targets != self.ignore_index
            safe_targets = torch.where(valid_mask, targets, torch.zeros_like(targets))
        else:
            valid_mask = torch.ones_like(targets, dtype=torch.bool)
            safe_targets = targets

        targets_one_hot = F.one_hot(safe_targets.long(), num_classes=self.num_classes)
        targets_one_hot = targets_one_hot.permute(0, 3, 1, 2).float()
        valid_mask = valid_mask.unsqueeze(1).float()
        probabilities = probabilities * valid_mask
        targets_one_hot = targets_one_hot * valid_mask

        dims = (0, 2, 3)
        targets_sum = torch.sum(targets_one_hot, dim=dims)

        present = targets_sum > 0
        safe_targets_sum = torch.where(
            present,
            targets_sum,
            torch.ones_like(targets_sum),
        )
        weights = present.to(probabilities.dtype) / (
            safe_targets_sum + self.smooth
        ).pow(2)
        weights = weights / weights.sum().clamp_min(self.smooth)

        intersection = torch.sum(probabilities * targets_one_hot, dim=dims)
        pred_sum = torch.sum(probabilities, dim=dims)

        weighted_intersection = torch.sum(weights * intersection)
        weighted_denominator = torch.sum(weights * (pred_sum + targets_sum))

        gdl = (2.0 * weighted_intersection + self.smooth) / (
            weighted_denominator + self.smooth
        )
        return 1.0 - gdl


class WeightedCE_GDL_Loss(nn.Module):
    def __init__(
        self,
        class_weights=None,
        num_classes=3,
        ce_loss_weight=1.0,
        gdl_loss_weight=1.0,
        smooth=1e-6,
        ignore_index=-100,
    ):
        super().__init__()
        self.ce_loss_weight = float(ce_loss_weight)
        self.gdl_loss_weight = float(gdl_loss_weight)

        ce_weights = (
            torch.tensor(class_weights, dtype=torch.float32)
            if class_weights is not None
            else None
        )

        self.register_buffer("ce_weights", ce_weights, persistent=True)

        self.cross_entropy = nn.CrossEntropyLoss(
            weight=self.ce_weights, ignore_index=ignore_index
        )
        self.gdl = GeneralizedDiceLoss(
            num_classes=num_classes, smooth=smooth, ignore_index=ignore_index
        )

    def get_extra_state(self):
        return {
            "ce_loss_weight": self.ce_loss_weight,
            "gdl_loss_weight": self.gdl_loss_weight,
            "num_classes": self.gdl.num_classes,
            "smooth": self.gdl.smooth,
            "ignore_index": self.gdl.ignore_index,
        }

    def set_extra_state(self, state):
        if not state:
            return

        self.ce_loss_weight = float(state["ce_loss_weight"])
        self.gdl_loss_weight = float(state["gdl_loss_weight"])
        self.gdl.num_classes = int(state["num_classes"])
        self.gdl.smooth = float(state["smooth"])
        self.gdl.ignore_index = state["ignore_index"]
        self.cross_entropy.ignore_index = state["ignore_index"]

    def forward(self, logits, targets):
        ce_loss = self.cross_entropy(logits, targets)
        gdl_loss = self.gdl(logits, targets)
        return self.ce_loss_weight * ce_loss + self.gdl_loss_weight * gdl_loss
```

## Mean Teacher consistency loss


```python
class SoftmaxMSEConsistencyLoss(nn.Module):
    def forward(self, student_logits, teacher_logits):
        if student_logits.shape != teacher_logits.shape:
            raise ValueError(
                "Student and Teacher logits must have identical shapes, "
                f"got {student_logits.shape} vs {teacher_logits.shape}."
            )

        student_prob = torch.softmax(student_logits, dim=1)
        teacher_prob = torch.softmax(teacher_logits.detach(), dim=1)
        return F.mse_loss(student_prob, teacher_prob, reduction="mean")
```

# Mean Teacher utils


```python
def create_teacher(student):
    teacher = copy.deepcopy(student)
    teacher.requires_grad_(False)

    # Swin-UNet uses LayerNorm, not running-stat BatchNorm.
    # Keep stochastic depth/dropout disabled for a stable EMA target.
    teacher.eval()
    return teacher


@torch.no_grad()
def update_ema_teacher(student, teacher, ema_decay, global_step):
    effective_decay = min(
        float(ema_decay),
        1.0 - 1.0 / (float(global_step) + 1.0),
    )

    student_parameters = dict(student.named_parameters())
    teacher_parameters = dict(teacher.named_parameters())

    if student_parameters.keys() != teacher_parameters.keys():
        raise RuntimeError("Student and Teacher parameter structures do not match.")

    for name, teacher_parameter in teacher_parameters.items():
        teacher_parameter.mul_(effective_decay).add_(
            student_parameters[name],
            alpha=1.0 - effective_decay,
        )


def sigmoid_rampup(current, rampup_length):
    rampup_length = float(rampup_length)
    if rampup_length <= 0:
        return 1.0

    current = float(np.clip(current, 0.0, rampup_length))
    phase = 1.0 - current / rampup_length
    return float(np.exp(-5.0 * phase * phase))


def get_consistency_weight(epoch, mt_config):
    maximum = float(mt_config["max_consistency_weight"])
    rampup = int(mt_config["consistency_rampup_epochs"])
    return maximum * sigmoid_rampup(
        current=max(epoch - 1, 0),
        rampup_length=rampup,
    )


def _next_cycled_batch(loader, iterator):
    try:
        batch = next(iterator)
    except StopIteration:
        iterator = iter(loader)
        batch = next(iterator)
    return batch, iterator
```

# Training and checkpoint utils


```python
def clone_state_dict_to_cpu(model):
    return {
        key: value.detach().cpu().clone() for key, value in model.state_dict().items()
    }


def component_type(component):
    if component is None:
        return None
    cls = component.__class__
    return f"{cls.__module__}.{cls.__qualname__}"


def empty_mean_teacher_history():
    return {
        "supervised_loss": [],
        "consistency_loss": [],
        "weighted_consistency_loss": [],
        "consistency_to_supervised_ratio": [],
        "total_loss": [],
        "train_dice": [],
        "student_val_loss": [],
        "student_val_dice": [],
        "teacher_val_loss": [],
        "teacher_val_dice": [],
        "consistency_weight": [],
        "learning_rate": [],
    }


def validate_resume_component(checkpoint, key, component):
    saved_type = checkpoint.get(f"{key}_type")
    current_type = component_type(component)

    if saved_type != current_type:
        raise ValueError(
            f"Cannot resume {key}: checkpoint uses {saved_type}, "
            f"but current notebook constructs {current_type}."
        )
```

```python
def save_mean_teacher_checkpoint(
    student,
    teacher,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    scheduler,
    epoch,
    global_step,
    best_loss,
    best_epoch,
    best_student_state_dict,
    best_teacher_state_dict,
    early_stopping_counter,
    history,
    config,
    mean_teacher_config,
    checkpoint_dir,
):
    checkpoint_dir = Path(checkpoint_dir)
    checkpoint_dir.mkdir(parents=True, exist_ok=True)

    timestamp = datetime.now().strftime(config.get("timestamp_format", "%y%m%d_%H%M%S"))
    path = checkpoint_dir / (
        f"mean_teacher_swin_unet_epoch_{epoch:03d}_{timestamp}.pth"
    )

    payload = {
        "training_method": "mean_teacher_swin_unet_softmax_mse",
        "epoch": int(epoch),
        "global_step": int(global_step),
        "student_type": component_type(student),
        "student_state_dict": student.state_dict(),
        "teacher_type": component_type(teacher),
        "teacher_state_dict": teacher.state_dict(),
        "supervised_criterion_type": component_type(supervised_criterion),
        "supervised_criterion_state_dict": supervised_criterion.state_dict(),
        "consistency_criterion_type": component_type(consistency_criterion),
        "consistency_criterion_state_dict": consistency_criterion.state_dict(),
        "optimizer_type": component_type(optimizer),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_type": component_type(scheduler),
        "scheduler_state_dict": (
            scheduler.state_dict() if scheduler is not None else None
        ),
        "best_loss": float(best_loss),
        "best_epoch": int(best_epoch),
        "best_student_state_dict": best_student_state_dict,
        "best_teacher_state_dict": best_teacher_state_dict,
        "early_stopping_counter": int(early_stopping_counter),
        "history": history,
        "config": {
            key: value for key, value in config.items() if not str(key).startswith("_")
        },
        "mean_teacher_config": copy.deepcopy(mean_teacher_config),
    }

    torch.save(payload, path)
    print(f"[Checkpoint] Saved: {path}")
    return path


def load_mean_teacher_checkpoint(
    checkpoint_path,
    student,
    teacher,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    scheduler,
    device,
):
    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=False,
    )

    if checkpoint.get("training_method") != "mean_teacher_swin_unet_softmax_mse":
        raise ValueError("The checkpoint is not a Swin-UNet Mean Teacher checkpoint.")

    for key, component in (
        ("student", student),
        ("teacher", teacher),
        ("supervised_criterion", supervised_criterion),
        ("consistency_criterion", consistency_criterion),
        ("optimizer", optimizer),
    ):
        validate_resume_component(checkpoint, key, component)

    if scheduler is not None:
        validate_resume_component(checkpoint, "scheduler", scheduler)

    student.load_state_dict(checkpoint["student_state_dict"])
    teacher.load_state_dict(checkpoint["teacher_state_dict"])
    supervised_criterion.load_state_dict(checkpoint["supervised_criterion_state_dict"])
    consistency_criterion.load_state_dict(
        checkpoint["consistency_criterion_state_dict"]
    )
    optimizer.load_state_dict(checkpoint["optimizer_state_dict"])

    if scheduler is not None and checkpoint.get("scheduler_state_dict") is not None:
        scheduler.load_state_dict(checkpoint["scheduler_state_dict"])

    teacher.requires_grad_(False)
    teacher.eval()
    return checkpoint


def save_best_mean_teacher(
    student,
    teacher,
    best_student_state_dict,
    best_teacher_state_dict,
    best_loss,
    best_epoch,
    history,
    config,
    mean_teacher_config,
):
    model_dir = Path(config["model_dir"])
    model_dir.mkdir(parents=True, exist_ok=True)

    path = model_dir / (f"best_mean_teacher_swin_unet_epoch_{best_epoch}.pth")

    torch.save(
        {
            "training_method": "mean_teacher_swin_unet_softmax_mse",
            "student_type": component_type(student),
            "student_state_dict": best_student_state_dict,
            "teacher_type": component_type(teacher),
            "teacher_state_dict": best_teacher_state_dict,
            "best_loss": float(best_loss),
            "best_epoch": int(best_epoch),
            "history": history,
            "config": {
                key: value
                for key, value in config.items()
                if not str(key).startswith("_")
            },
            "mean_teacher_config": copy.deepcopy(mean_teacher_config),
            "class_names": CLASS_NAMES,
            "class_colors": CLASS_COLORS,
        },
        path,
    )

    print(f"[Best model] Saved: {path}")
    return path
```

## One epoch


```python
def train_mean_teacher_one_epoch(
    student,
    teacher,
    labeled_loader,
    unlabeled_loader,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    device,
    epoch,
    global_step,
    num_classes,
    mean_teacher_config,
):
    student.train()
    teacher.eval()

    consistency_weight = get_consistency_weight(epoch, mean_teacher_config)
    ema_decay = float(mean_teacher_config["ema_decay"])

    steps = max(len(labeled_loader), len(unlabeled_loader))
    labeled_iterator = iter(labeled_loader)
    unlabeled_iterator = iter(unlabeled_loader)

    supervised_sum = 0.0
    consistency_sum = 0.0
    weighted_consistency_sum = 0.0
    consistency_ratio_sum = 0.0
    total_sum = 0.0
    dice_sum = 0.0
    labeled_samples = 0

    for _ in range(steps):
        labeled_batch, labeled_iterator = _next_cycled_batch(
            labeled_loader, labeled_iterator
        )
        unlabeled_batch, unlabeled_iterator = _next_cycled_batch(
            unlabeled_loader, unlabeled_iterator
        )

        labeled_weak = labeled_batch["weak_image"].to(device, non_blocking=True)
        labeled_strong = labeled_batch["strong_image"].to(device, non_blocking=True)
        labeled_masks = labeled_batch["mask"].to(device, non_blocking=True).long()

        unlabeled_weak = unlabeled_batch["weak_image"].to(device, non_blocking=True)
        unlabeled_strong = unlabeled_batch["strong_image"].to(device, non_blocking=True)

        num_labeled = labeled_strong.size(0)

        # One Student forward and one Teacher forward per step.
        student_input = torch.cat(
            [labeled_strong, unlabeled_strong],
            dim=0,
        )
        teacher_input = torch.cat(
            [labeled_weak, unlabeled_weak],
            dim=0,
        )

        optimizer.zero_grad(set_to_none=True)

        student_logits = student(student_input)

        with torch.no_grad():
            teacher_logits = teacher(teacher_input)

        supervised_loss = supervised_criterion(
            student_logits[:num_labeled],
            labeled_masks,
        )

        # Classic Mean Teacher: consistency uses labeled + unlabeled inputs.
        consistency_loss = consistency_criterion(
            student_logits,
            teacher_logits,
        )

        weighted_consistency_loss = consistency_weight * consistency_loss
        total_loss = supervised_loss + weighted_consistency_loss

        consistency_to_supervised_ratio = (
            weighted_consistency_loss.detach()
            / supervised_loss.detach().clamp_min(1e-8)
        )

        total_loss.backward()
        optimizer.step()

        global_step += 1
        update_ema_teacher(
            student,
            teacher,
            ema_decay=ema_decay,
            global_step=global_step,
        )

        with torch.no_grad():
            dice = multiclass_dice_score(
                student_logits[:num_labeled],
                labeled_masks,
                num_classes=num_classes,
                include_background=False,
            )

        supervised_sum += supervised_loss.item()
        consistency_sum += consistency_loss.item()
        weighted_consistency_sum += weighted_consistency_loss.item()
        consistency_ratio_sum += consistency_to_supervised_ratio.item()
        total_sum += total_loss.item()
        dice_sum += dice.item() * num_labeled
        labeled_samples += num_labeled

    metrics = {
        "supervised_loss": supervised_sum / max(steps, 1),
        "consistency_loss": consistency_sum / max(steps, 1),
        "weighted_consistency_loss": weighted_consistency_sum / max(steps, 1),
        "consistency_to_supervised_ratio": consistency_ratio_sum / max(steps, 1),
        "total_loss": total_sum / max(steps, 1),
        "train_dice": dice_sum / max(labeled_samples, 1),
        "consistency_weight": consistency_weight,
    }

    return metrics, global_step


@torch.no_grad()
def validate_one_epoch(
    model,
    loader,
    criterion,
    device,
    num_classes=3,
):
    model.eval()

    running_loss = 0.0
    running_dice = 0.0
    total_samples = 0

    for batch in loader:
        images = batch["image"].to(device, non_blocking=True)
        masks = batch["mask"].to(device, non_blocking=True).long()

        logits = model(images)
        loss = criterion(logits, masks)
        batch_size = images.size(0)

        dice = multiclass_dice_score(
            logits,
            masks,
            num_classes=num_classes,
            include_background=False,
        )

        running_loss += loss.item() * batch_size
        running_dice += dice.item() * batch_size
        total_samples += batch_size

    return (
        running_loss / max(total_samples, 1),
        running_dice / max(total_samples, 1),
    )
```

## Training loop


```python
def train_mean_teacher(
    student,
    teacher,
    labeled_loader,
    unlabeled_loader,
    val_loader,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    scheduler,
    config,
    mean_teacher_config,
    checkpoint_path=None,
):
    device = torch.device(config["device"])
    target_epoch = int(config["epochs"])
    num_classes = int(config["num_classes"])
    save_every = int(config.get("save_every", 0))

    early_stopping_enabled = bool(config.get("early_stopping", False))
    patience = int(config.get("early_stopping_patience", 10))
    min_delta = float(config.get("early_stopping_min_delta", 0.0))

    if patience < 1:
        raise ValueError("early_stopping_patience must be >= 1.")
    if min_delta < 0:
        raise ValueError("early_stopping_min_delta must be >= 0.")

    history = empty_mean_teacher_history()
    start_epoch = 1
    global_step = 0
    best_loss = float("inf")
    best_epoch = 0
    best_student_state_dict = None
    best_teacher_state_dict = None
    early_stopping_counter = 0

    if checkpoint_path is not None:
        checkpoint = load_mean_teacher_checkpoint(
            checkpoint_path=checkpoint_path,
            student=student,
            teacher=teacher,
            supervised_criterion=supervised_criterion,
            consistency_criterion=consistency_criterion,
            optimizer=optimizer,
            scheduler=scheduler,
            device=device,
        )

        start_epoch = int(checkpoint["epoch"]) + 1
        global_step = int(checkpoint.get("global_step", 0))
        best_loss = float(checkpoint.get("best_loss", float("inf")))
        best_epoch = int(checkpoint.get("best_epoch", 0))
        best_student_state_dict = checkpoint.get("best_student_state_dict")
        best_teacher_state_dict = checkpoint.get("best_teacher_state_dict")
        early_stopping_counter = int(checkpoint.get("early_stopping_counter", 0))

        history = checkpoint.get(
            "history",
            empty_mean_teacher_history(),
        )

        completed_epochs = start_epoch - 1
        for key in empty_mean_teacher_history().keys():
            values = history.setdefault(key, [])
            if len(values) < completed_epochs:
                values.extend([float("nan")] * (completed_epochs - len(values)))

        print(
            f"[Resume] epoch {start_epoch - 1} -> {target_epoch} | "
            f"global_step={global_step}"
        )
    else:
        print(f"[Train] epoch 1 -> {target_epoch}")

    if start_epoch > target_epoch:
        print(
            f"[Warning] No epoch to train: "
            f"start={start_epoch}, target={target_epoch}"
        )
        return student, teacher, history

    last_epoch = start_epoch - 1

    for epoch in range(start_epoch, target_epoch + 1):
        last_epoch = epoch

        train_metrics, global_step = train_mean_teacher_one_epoch(
            student=student,
            teacher=teacher,
            labeled_loader=labeled_loader,
            unlabeled_loader=unlabeled_loader,
            supervised_criterion=supervised_criterion,
            consistency_criterion=consistency_criterion,
            optimizer=optimizer,
            device=device,
            epoch=epoch,
            global_step=global_step,
            num_classes=num_classes,
            mean_teacher_config=mean_teacher_config,
        )

        student_val_loss, student_val_dice = validate_one_epoch(
            student,
            val_loader,
            supervised_criterion,
            device,
            num_classes,
        )

        teacher_val_loss, teacher_val_dice = validate_one_epoch(
            teacher,
            val_loader,
            supervised_criterion,
            device,
            num_classes,
        )

        improved = teacher_val_loss < (best_loss - min_delta)

        if improved:
            best_loss = teacher_val_loss
            best_epoch = epoch
            best_student_state_dict = clone_state_dict_to_cpu(student)
            best_teacher_state_dict = clone_state_dict_to_cpu(teacher)
            early_stopping_counter = 0
        elif early_stopping_enabled:
            early_stopping_counter += 1

        if scheduler is not None:
            if isinstance(
                scheduler,
                torch.optim.lr_scheduler.ReduceLROnPlateau,
            ):
                scheduler.step(teacher_val_loss)
            else:
                scheduler.step()

        current_lr = optimizer.param_groups[0]["lr"]

        for key in (
            "supervised_loss",
            "consistency_loss",
            "weighted_consistency_loss",
            "consistency_to_supervised_ratio",
            "total_loss",
            "train_dice",
            "consistency_weight",
        ):
            history[key].append(train_metrics[key])

        history["student_val_loss"].append(student_val_loss)
        history["student_val_dice"].append(student_val_dice)
        history["teacher_val_loss"].append(teacher_val_loss)
        history["teacher_val_dice"].append(teacher_val_dice)
        history["learning_rate"].append(current_lr)

        early_stop_text = ""
        if early_stopping_enabled:
            early_stop_text = f" | early_stop={early_stopping_counter}/{patience}"

        print(
            f"Epoch {epoch:02d}/{target_epoch:02d} | "
            f"sup={train_metrics['supervised_loss']:.5f} | "
            f"cons={train_metrics['consistency_loss']:.6f} | "
            f"w_cons={train_metrics['weighted_consistency_loss']:.6f} | "
            f"w_cons/sup={train_metrics['consistency_to_supervised_ratio']:.4f} | "
            f"lambda_c={train_metrics['consistency_weight']:.4f} | "
            f"train_dice={train_metrics['train_dice']:.4f} | "
            f"student_val_dice={student_val_dice:.4f} | "
            f"teacher_val_dice={teacher_val_dice:.4f} | "
            f"lr={current_lr:.7f} | "
            f"best_teacher_val_loss={best_loss:.5f} @ {best_epoch}"
            f"{early_stop_text}"
        )

        if save_every > 0 and epoch % save_every == 0:
            save_mean_teacher_checkpoint(
                student=student,
                teacher=teacher,
                supervised_criterion=supervised_criterion,
                consistency_criterion=consistency_criterion,
                optimizer=optimizer,
                scheduler=scheduler,
                epoch=epoch,
                global_step=global_step,
                best_loss=best_loss,
                best_epoch=best_epoch,
                best_student_state_dict=best_student_state_dict,
                best_teacher_state_dict=best_teacher_state_dict,
                early_stopping_counter=early_stopping_counter,
                history=history,
                config=config,
                mean_teacher_config=mean_teacher_config,
                checkpoint_dir=config["checkpoint_dir"],
            )

        if early_stopping_enabled and early_stopping_counter >= patience:
            print(
                "[Early stopping] "
                f"No Teacher validation-loss improvement for {patience} epochs. "
                f"Best epoch: {best_epoch}."
            )
            break

    if best_student_state_dict is None or best_teacher_state_dict is None:
        best_student_state_dict = clone_state_dict_to_cpu(student)
        best_teacher_state_dict = clone_state_dict_to_cpu(teacher)
        best_epoch = last_epoch
        best_loss = teacher_val_loss

    save_best_mean_teacher(
        student=student,
        teacher=teacher,
        best_student_state_dict=best_student_state_dict,
        best_teacher_state_dict=best_teacher_state_dict,
        best_loss=best_loss,
        best_epoch=best_epoch,
        history=history,
        config=config,
        mean_teacher_config=mean_teacher_config,
    )

    student.load_state_dict(best_student_state_dict)
    teacher.load_state_dict(best_teacher_state_dict)

    student.to(device)
    teacher.to(device).requires_grad_(False).eval()

    return student, teacher, history
```

```python
@torch.no_grad()
def evaluate_model(
    model,
    loader,
    device,
    num_classes=3,
    class_names=None,
):
    """
    Evaluate model on a dataset and compute metrics per class.

    Args:
        model: model to evaluate
        loader: DataLoader of the evaluation dataset
        device: device used for evaluation
        num_classes: number of segmentation classes
        class_names: dictionary mapping class IDs to class names

    Returns:
        DataFrame containing:
            - metrics for each class
            - macro_foreground_only:
                average of foreground classes only
            - macro_all_classes:
                average of all classes including background
    """

    class_names = class_names or {i: f"class_{i}" for i in range(num_classes)}

    tp = np.zeros(
        num_classes,
        dtype=np.float64,
    )

    fp = np.zeros(
        num_classes,
        dtype=np.float64,
    )

    fn = np.zeros(
        num_classes,
        dtype=np.float64,
    )

    hd95_values = {class_id: [] for class_id in range(num_classes)}

    model.eval()

    for batch in loader:
        images = batch["image"].to(
            device,
            non_blocking=True,
        )

        masks = (
            batch["mask"]
            .to(
                device,
                non_blocking=True,
            )
            .long()
        )

        logits = model(images)

        preds = torch.argmax(
            logits,
            dim=1,
        )

        preds_np = preds.cpu().numpy()
        masks_np = masks.cpu().numpy()

        for class_id in range(num_classes):
            pred_c = preds_np == class_id
            true_c = masks_np == class_id

            tp[class_id] += np.logical_and(
                pred_c,
                true_c,
            ).sum()

            fp[class_id] += np.logical_and(
                pred_c,
                ~true_c,
            ).sum()

            fn[class_id] += np.logical_and(
                ~pred_c,
                true_c,
            ).sum()

            for sample_idx in range(preds_np.shape[0]):
                hd95_values[class_id].append(
                    hd95_binary(
                        pred_c[sample_idx],
                        true_c[sample_idx],
                    )
                )

    rows = []

    for class_id in range(num_classes):
        dice = _safe_divide(
            2.0 * tp[class_id],
            (2.0 * tp[class_id] + fp[class_id] + fn[class_id]),
        )

        iou = _safe_divide(
            tp[class_id],
            (tp[class_id] + fp[class_id] + fn[class_id]),
        )

        precision = _safe_divide(
            tp[class_id],
            (tp[class_id] + fp[class_id]),
        )

        recall = _safe_divide(
            tp[class_id],
            (tp[class_id] + fn[class_id]),
        )

        hd_values = np.asarray(
            hd95_values[class_id],
            dtype=np.float64,
        )

        hd95 = float(np.nanmean(hd_values)) if np.isfinite(hd_values).any() else np.nan

        rows.append(
            {
                "class_id": class_id,
                "class_name": class_names.get(
                    class_id,
                    f"class_{class_id}",
                ),
                "dice": dice,
                "iou": iou,
                "hd95": hd95,
                "precision": precision,
                "recall": recall,
            }
        )

    result = pd.DataFrame(rows)

    # Macro foreground only
    foreground = result[result["class_id"].isin(range(1, num_classes))]

    macro_fg_row = {
        "class_id": "macro_fg",
        "class_name": "macro_foreground_only",
        "dice": foreground["dice"].mean(skipna=True),
        "iou": foreground["iou"].mean(skipna=True),
        "hd95": foreground["hd95"].mean(skipna=True),
        "precision": foreground["precision"].mean(skipna=True),
        "recall": foreground["recall"].mean(skipna=True),
    }

    # Macro all classes includes background
    all_classes = result[result["class_id"].isin(range(num_classes))]

    macro_all_row = {
        "class_id": "macro_all",
        "class_name": "macro_all_classes",
        "dice": all_classes["dice"].mean(skipna=True),
        "iou": all_classes["iou"].mean(skipna=True),
        "hd95": all_classes["hd95"].mean(skipna=True),
        "precision": all_classes["precision"].mean(skipna=True),
        "recall": all_classes["recall"].mean(skipna=True),
    }

    return pd.concat(
        [
            result,
            pd.DataFrame(
                [
                    macro_fg_row,
                    macro_all_row,
                ]
            ),
        ],
        ignore_index=True,
    )
```

## Setup


```python
if RESUME_CHECKPOINT is None:
    CONFIG = copy.deepcopy(DEFAULT_CONFIG)
    MT_CONFIG = copy.deepcopy(DEFAULT_MEAN_TEACHER_CONFIG)
else:
    resume_path = Path(RESUME_CHECKPOINT)

    if not resume_path.exists():
        raise FileNotFoundError(f"Resume checkpoint not found: {resume_path}")

    checkpoint_metadata = torch.load(
        resume_path,
        map_location="cpu",
        weights_only=False,
    )

    if (
        checkpoint_metadata.get("training_method")
        != "mean_teacher_swin_unet_softmax_mse"
    ):
        raise ValueError(
            "The selected checkpoint is not a Swin-UNet Mean Teacher checkpoint."
        )

    checkpoint_config = checkpoint_metadata.get("config")
    checkpoint_mt_config = checkpoint_metadata.get("mean_teacher_config")

    if not isinstance(checkpoint_config, dict):
        raise KeyError("Checkpoint does not contain a valid config.")
    if not isinstance(checkpoint_mt_config, dict):
        raise KeyError("Checkpoint does not contain a valid mean_teacher_config.")

    CONFIG = copy.deepcopy(checkpoint_config)
    MT_CONFIG = copy.deepcopy(checkpoint_mt_config)

# Notebook controls the final target epoch and save locations.
CONFIG["epochs"] = int(DEFAULT_CONFIG["epochs"])
CONFIG["checkpoint_dir"] = str(CHECKPOINT_DIR)
CONFIG["model_dir"] = str(MODEL_DIR)

for key, value in DEFAULT_CONFIG.items():
    CONFIG.setdefault(key, copy.deepcopy(value))
for key, value in DEFAULT_MEAN_TEACHER_CONFIG.items():
    MT_CONFIG.setdefault(key, copy.deepcopy(value))

print("Config source              :", "Checkpoint" if RESUME_CHECKPOINT else "Default")
print("Device                     :", CONFIG["device"])
print("Architecture               : Swin-UNet Mean Teacher")
print(
    "Input size                 :", f"{CONFIG['image_height']}x{CONFIG['image_width']}"
)
print("Train subset               :", CONFIG["train_subset"])
print("Labeled batch              :", CONFIG["labeled_batch_size"])
print("Unlabeled batch            :", CONFIG["unlabeled_batch_size"])
print("Target epochs              :", CONFIG["epochs"])
print("EMA decay                  :", MT_CONFIG["ema_decay"])
print("Max consistency weight     :", MT_CONFIG["max_consistency_weight"])
print("Consistency ramp-up epochs :", MT_CONFIG["consistency_rampup_epochs"])
```

## Download


```python
path = kagglehub.dataset_download("avcuongy/bkai-igh-neopolyp-small-sample")
print(path)

DATA_DIR = Path(os.path.join(path, "data"))
print(DATA_DIR)

SPLIT_DIR = DATA_DIR / "sample"
TRAIN_CSV_PATH = SPLIT_DIR / "train.csv"
VALID_CSV_PATH = SPLIT_DIR / "valid.csv"
TEST_CSV_PATH = SPLIT_DIR / "test.csv"

train_df = pd.read_csv(TRAIN_CSV_PATH)
valid_df = pd.read_csv(VALID_CSV_PATH)
test_df = pd.read_csv(TEST_CSV_PATH)
print("Train:", train_df.shape)
print("Valid:", valid_df.shape)
print("Test :", test_df.shape)
print()
print(
    "Train columns:",
    train_df.columns.tolist(),
)
print(
    "Valid columns:",
    valid_df.columns.tolist(),
)
print(
    "Test columns:",
    test_df.columns.tolist(),
)
```

```python
TRAIN_SUBSET_FLAG_MAP = {
    "half": "labeled_1_2",
    "quarter": "labeled_1_4",
}

train_subset = str(CONFIG["train_subset"]).strip().lower()

if train_subset not in TRAIN_SUBSET_FLAG_MAP:
    raise ValueError(
        "Mean Teacher requires an unlabeled pool. "
        "CONFIG['train_subset'] must be 'half' or 'quarter'."
    )

flag_column = TRAIN_SUBSET_FLAG_MAP[train_subset]

if flag_column not in train_df.columns:
    raise KeyError(f"Missing required split column: {flag_column}")

flag_values = pd.to_numeric(
    train_df[flag_column],
    errors="coerce",
)

if flag_values.isna().any():
    raise ValueError(f"Column '{flag_column}' contains missing/non-numeric values.")

unique_flags = set(flag_values.astype(int).unique().tolist())

if not unique_flags.issubset({0, 1}):
    raise ValueError(
        f"Column '{flag_column}' must contain only 0/1. "
        f"Found: {sorted(unique_flags)}"
    )

labeled_train_df = train_df[flag_values.eq(1)].copy().reset_index(drop=True)

unlabeled_train_df = train_df[flag_values.eq(0)].copy().reset_index(drop=True)

if labeled_train_df.empty:
    raise ValueError("The labeled training split is empty.")
if unlabeled_train_df.empty:
    raise ValueError("The unlabeled training split is empty.")

print("Split column       :", flag_column)
print("Train total        :", len(train_df))
print("Labeled training   :", len(labeled_train_df))
print("Unlabeled training :", len(unlabeled_train_df))
print("Validation         :", len(valid_df))
print("Test               :", len(test_df))
```

## DataLoaders


```python
augmentation_seed = int(CONFIG["seed"])

labeled_shared_geometry = build_labeled_shared_geometry(
    CONFIG["image_height"],
    CONFIG["image_width"],
    seed=augmentation_seed + 0,
)

unlabeled_shared_geometry = build_unlabeled_shared_geometry(
    CONFIG["image_height"],
    CONFIG["image_width"],
    seed=augmentation_seed + 1,
)

labeled_weak_transform = build_weak_transform(
    seed=augmentation_seed + 2,
)
labeled_strong_transform = build_labeled_strong_transform(
    seed=augmentation_seed + 3,
)

unlabeled_weak_transform = build_weak_transform(
    seed=augmentation_seed + 4,
)
unlabeled_strong_transform = build_unlabeled_strong_transform(
    seed=augmentation_seed + 5,
)

eval_transform = build_eval_transform(
    CONFIG["image_height"],
    CONFIG["image_width"],
    seed=augmentation_seed + 6,
)

labeled_dataset = MeanTeacherLabeledDataset(
    dataframe=labeled_train_df,
    image_dir=DATA_DIR,
    mask_dir=DATA_DIR,
    shared_geometry=labeled_shared_geometry,
    weak_transform=labeled_weak_transform,
    strong_transform=labeled_strong_transform,
    image_col=CONFIG["image_col"],
    mask_col=CONFIG["mask_col"],
)

unlabeled_dataset = MeanTeacherUnlabeledDataset(
    dataframe=unlabeled_train_df,
    image_dir=DATA_DIR,
    shared_geometry=unlabeled_shared_geometry,
    weak_transform=unlabeled_weak_transform,
    strong_transform=unlabeled_strong_transform,
    image_col=CONFIG["image_col"],
)

valid_dataset = PolypSegmentationDataset(
    dataframe=valid_df,
    image_dir=DATA_DIR,
    mask_dir=DATA_DIR,
    transform=eval_transform,
    image_col=CONFIG["image_col"],
    mask_col=CONFIG["mask_col"],
    has_mask=True,
)

test_dataset = PolypSegmentationDataset(
    dataframe=test_df,
    image_dir=DATA_DIR,
    mask_dir=DATA_DIR,
    transform=eval_transform,
    image_col=CONFIG["image_col"],
    mask_col=CONFIG["mask_col"],
    has_mask=True,
)

# Fail before training if columns/paths/masks are wrong.
validate_labeled_dataset(
    labeled_dataset,
    "labeled_train",
    max_samples=8,
)
validate_unlabeled_dataset(
    unlabeled_dataset,
    "unlabeled_train",
    max_samples=8,
)
validate_labeled_dataset(
    valid_dataset,
    "validation",
    max_samples=8,
)
validate_labeled_dataset(
    test_dataset,
    "test",
    max_samples=8,
)


labeled_generator = torch.Generator()
labeled_generator.manual_seed(CONFIG["seed"])

unlabeled_generator = torch.Generator()
unlabeled_generator.manual_seed(CONFIG["seed"] + 1)

common_loader_args = {
    "num_workers": CONFIG["num_workers"],
    "pin_memory": torch.cuda.is_available(),
    "worker_init_fn": seed_worker,
    "persistent_workers": CONFIG["num_workers"] > 0,
}

labeled_loader = DataLoader(
    labeled_dataset,
    batch_size=CONFIG["labeled_batch_size"],
    shuffle=True,
    generator=labeled_generator,
    **common_loader_args,
)

unlabeled_loader = DataLoader(
    unlabeled_dataset,
    batch_size=CONFIG["unlabeled_batch_size"],
    shuffle=True,
    generator=unlabeled_generator,
    **common_loader_args,
)

valid_loader = DataLoader(
    valid_dataset,
    batch_size=CONFIG["eval_batch_size"],
    shuffle=False,
    **common_loader_args,
)

test_loader = DataLoader(
    test_dataset,
    batch_size=CONFIG["eval_batch_size"],
    shuffle=False,
    **common_loader_args,
)
```

```python
print("Labeled samples      :", len(labeled_dataset))
print("Unlabeled samples    :", len(unlabeled_dataset))
print("Validation samples   :", len(valid_dataset))
print("Test samples         :", len(test_dataset))
print("Labeled batches      :", len(labeled_loader))
print("Unlabeled batches    :", len(unlabeled_loader))
print(
    "Student batch / step :",
    CONFIG["labeled_batch_size"] + CONFIG["unlabeled_batch_size"],
)
```

## Model


```python
device = torch.device(CONFIG["device"])
image_height = int(CONFIG["image_height"])
image_width = int(CONFIG["image_width"])

if image_height != image_width:
    raise ValueError(
        "This SwinUNet implementation requires square input. "
        f"Got {image_height}x{image_width}."
    )

configured_pretrained_path = CONFIG.get("pretrained_path")
pretrained_path = (
    str(configured_pretrained_path).strip()
    if configured_pretrained_path is not None
    else None
)
pretrained_path = pretrained_path or None

# Resume loads Student/Teacher states, so ImageNet initialization is not repeated.
if RESUME_CHECKPOINT is not None:
    pretrained_path = None
elif pretrained_path is not None and not Path(pretrained_path).exists():
    raise FileNotFoundError(
        "Swin pretrained checkpoint not found: "
        f"{pretrained_path}. Set CONFIG['pretrained_path']=None "
        "to train from scratch."
    )

student = SwinUNet(
    img_size=image_height,
    patch_size=CONFIG["patch_size"],
    in_channels=CONFIG["in_channels"],
    num_classes=CONFIG["num_classes"],
    embed_dim=CONFIG["embed_dim"],
    depths=CONFIG["depths"],
    num_heads=CONFIG["num_heads"],
    window_size=CONFIG["window_size"],
    mlp_ratio=CONFIG["mlp_ratio"],
    qkv_bias=CONFIG["qkv_bias"],
    drop_rate=CONFIG["drop_rate"],
    attn_drop_rate=CONFIG["attn_drop_rate"],
    drop_path_rate=CONFIG["drop_path_rate"],
    ape=CONFIG["ape"],
    patch_norm=CONFIG["patch_norm"],
    use_checkpoint=CONFIG["use_checkpoint"],
    pretrained_path=pretrained_path,
    pretrained_decoder=CONFIG["pretrained_decoder"],
).to(device)

freeze_encoder = bool(CONFIG.get("freeze_encoder", False))

for parameter in student.encoder.parameters():
    parameter.requires_grad = not freeze_encoder

teacher = create_teacher(student).to(device)
```

```python
supervised_criterion = WeightedCE_GDL_Loss(
    class_weights=[0.2, 2.0, 1.0],
    num_classes=CONFIG["num_classes"],
    ce_loss_weight=1.0,
    gdl_loss_weight=2.0,
    smooth=1e-8,
    ignore_index=-100,
).to(device)

consistency_criterion = SoftmaxMSEConsistencyLoss().to(device)

trainable_parameters = [
    parameter for parameter in student.parameters() if parameter.requires_grad
]

optimizer = torch.optim.AdamW(
    trainable_parameters,
    lr=3e-4,
    weight_decay=1e-4,
)

scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(
    optimizer,
    T_max=CONFIG["epochs"],
    eta_min=1e-6,
)
```

```python
print("Student                 :", student.__class__.__name__)
print("Teacher                 :", teacher.__class__.__name__)
print(
    "Teacher trainable params:",
    sum(p.numel() for p in teacher.parameters() if p.requires_grad),
)
print("Pretrained path         :", pretrained_path)
print("Freeze encoder          :", freeze_encoder)
print("Supervised criterion  :", supervised_criterion.__class__.__name__)
print("Consistency criterion :", consistency_criterion.__class__.__name__)
print("Optimizer             :", optimizer.__class__.__name__)
print("Scheduler             :", scheduler.__class__.__name__)
print("Mean Teacher settings :", MT_CONFIG)
```

## Train


```python
student, teacher, history = train_mean_teacher(
    student=student,
    teacher=teacher,
    labeled_loader=labeled_loader,
    unlabeled_loader=unlabeled_loader,
    val_loader=valid_loader,
    supervised_criterion=supervised_criterion,
    consistency_criterion=consistency_criterion,
    optimizer=optimizer,
    scheduler=scheduler,
    config=CONFIG,
    mean_teacher_config=MT_CONFIG,
    checkpoint_path=RESUME_CHECKPOINT,
)
```
