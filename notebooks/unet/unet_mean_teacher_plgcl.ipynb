# Mean Teacher + PLGCL (Student patch embeddings)

Hai notebook để so sánh **loss InfoNCE và PLGCL trên Student**. Chạy từng notebook
trong kernel mới, giữ cùng `SEED`, split và ngân sách huấn luyện. Dùng patch encoder
ở cả hai bản để tránh so sánh đồng thời cả loss lẫn vị trí lấy đặc trưng.

Two matched experiments, based on `unet_mean_teacher.ipynb`. Only the contrastive
objective changes between the notebooks. Run each in a fresh kernel with the same
seed, split, data, hardware and training settings. Original baseline files are unchanged.

- Shared: ResNet U-Net, Weighted CE + GDL, softmax-MSE consistency, EMA Teacher,
  augmentation, Student patch encoder/head, patch sampling, temperature and schedules.
- Student generates detached class-confidence maps for patch selection; labeled
  images use ground truth. Teacher supplies consistency targets, **not contrastive keys**.
- Both query and keys are Student embeddings; gradients reach the shared encoder and
  projection head. The decoder learns from segmentation/consistency losses.
- The PLGCL branch implements the Gaussian mean/covariance objective in Eq. (5),
  rather than renaming supervised InfoNCE. Its covariance quadratic is evaluated
  exactly from projected centered samples without allocating D x D covariance matrices.
- Controlled adaptation, not an exact CVPR reproduction: retain the existing MSE/GDL,
  use RGB attended patches, confidence/purity filtering, a bounded grid sampler,
  within-class entropy-nearest positives, equal-count other-class negatives, normalized
  embeddings, batch-estimated population covariance and a shorter warm-up.
  The paper's sampler wording is ambiguous about leftover same-class patches; here
  they are excluded, never relabeled as negatives. Patch BatchNorm uses running
  statistics so extra patch passes do not alter the full-image BatchNorm buffers.
  The sampling rule is identical, but actual pseudo-labels and sampled patches may
  diverge as the two Students learn. This is a comparison of objectives within the
  same pipeline, not a claim that every later batch has identical contrastive keys.

Reference: [Basak & Yin, CVPR 2023, Sections 3.1–3.3 / Eqs. (4)–(5)]
(https://openaccess.thecvf.com/content/CVPR2023/papers/Basak_Pseudo-Label_Guided_Contrastive_Learning_for_Semi-Supervised_Medical_Image_Segmentation_CVPR_2023_paper.pdf).

For an equal training budget, both new notebooks disable early stopping and retain
50 epochs. Their warm-up is 5 epochs, followed by a 10-epoch linear ramp. These are
starting settings, not tuned results. If comparing to plain Mean Teacher, rerun that
baseline with the same fixed epoch budget. Select checkpoints using **Teacher
validation foreground Dice**, never test metrics. Run at least seeds 42, 43, 44 and
report per-class and foreground Dice/IoU mean ± std; tune loss weights on validation
with equal search budgets. Training loss values of different objectives are not comparable.

Outputs/checkpoints are isolated by method, split and seed. Checkpoints include the
projection head, optimizer, schedules/config and contrastive history. Resume restores
training state but is not promised bitwise-identical after restarting worker augmentation
streams. Do not resume another method's checkpoint. `RESUME_CHECKPOINT` is for trusted
local checkpoints only. No full-data training result has been produced with these files.

`LOCAL_DATA_DIR` accepts an existing dataset root containing `sample/train.csv`,
`sample/valid.csv`, `sample/test.csv` and the image/mask paths referenced in those CSVs.
Leave it as `None` to use the same Kaggle dataset as the baseline. At the end, the
notebook exports metrics/configuration and compares both methods when both results
exist. Compare test metrics only after selecting settings/checkpoints on validation.

```python
import cv2
import albumentations as A
from albumentations.pytorch import ToTensorV2

# SEED is defined once in the reproducibility cell above.


def build_labeled_shared_geometry(height=224, width=224, seed=SEED):
    """Geometric-only view applied jointly to image and mask."""
    return A.Compose(
        [
            A.Resize(height=height, width=width),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.5),
            A.RandomRotate90(p=0.5),
            A.Affine(
                translate_percent={"x": (-0.0625, 0.0625), "y": (-0.0625, 0.0625)},
                scale=(0.85, 1.15),
                rotate=(-45, 45),
                border_mode=cv2.BORDER_CONSTANT,  # Sửa 'mode' thành 'border_mode' ở đây
                p=0.5,
            ),
            A.OneOf(
                [
                    A.ElasticTransform(
                        alpha=120,
                        sigma=120 * 0.05,
                        p=0.5,
                    ),
                    A.GridDistortion(p=0.5),
                    A.OpticalDistortion(
                        distort_limit=(-1.0, 1.0),
                        p=0.5,
                    ),
                ],
                p=0.4,
            ),
        ],
        seed=int(seed),
        strict=True,
    )


def build_labeled_weak_transform(seed=SEED):
    """Teacher view of labeled images: normalize only, no photometric noise."""
    return A.Compose(
        [
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )


def build_labeled_strong_transform(seed=SEED):
    """Student view: photometric-only perturbations after shared geometry."""
    return A.Compose(
        [
            A.OneOf(
                [
                    A.RandomBrightnessContrast(
                        brightness_limit=(-0.10, 0.10),
                        contrast_limit=(-0.10, 0.10),
                        p=0.5,
                    ),
                    A.HueSaturationValue(
                        hue_shift_limit=(-10, 10),
                        sat_shift_limit=(-15, 15),
                        val_shift_limit=(-10, 10),
                        p=0.5,
                    ),
                ],
                p=0.3,
            ),
            A.OneOf(
                [
                    A.GaussNoise(p=0.5),
                    A.GaussianBlur(blur_limit=(3, 5), p=0.5),
                ],
                p=0.2,
            ),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )


def build_unlabeled_shared_geometry(height=224, width=224, seed=SEED):
    """One geometric view shared by Teacher and Student to preserve alignment."""
    return A.Compose(
        [
            A.Resize(height=height, width=width),
            A.HorizontalFlip(p=0.5),
            A.VerticalFlip(p=0.5),
            A.RandomRotate90(p=0.5),
            A.Affine(
                translate_percent={"x": (-0.0625, 0.0625), "y": (-0.0625, 0.0625)},
                scale=(0.90, 1.10),
                rotate=(-20, 20),
                border_mode=cv2.BORDER_CONSTANT,
                p=0.5,
            ),
        ],
        seed=int(seed),
        strict=True,
    )


def build_unlabeled_weak_transform(seed=SEED):
    return A.Compose(
        [
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )


def build_unlabeled_strong_transform(seed=SEED):
    """Photometric-only perturbations keep Teacher/Student pixels aligned."""
    return A.Compose(
        [
            A.OneOf(
                [
                    A.RandomBrightnessContrast(
                        brightness_limit=(-0.20, 0.20),
                        contrast_limit=(-0.20, 0.20),
                        p=1.0,
                    ),
                    A.HueSaturationValue(
                        hue_shift_limit=(-15, 15),
                        sat_shift_limit=(-25, 25),
                        val_shift_limit=(-15, 15),
                        p=1.0,
                    ),
                ],
                p=0.8,
            ),
            A.OneOf(
                [
                    A.GaussNoise(p=1.0),
                    A.GaussianBlur(blur_limit=(3, 5), p=1.0),
                ],
                p=0.3,
            ),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )


def build_eval_transform(height=224, width=224, seed=SEED):
    return A.Compose(
        [
            A.Resize(height=height, width=width),
            A.Normalize(
                mean=(0.485, 0.456, 0.406),
                std=(0.229, 0.224, 0.225),
            ),
            ToTensorV2(),
        ],
        seed=int(seed),
        strict=True,
    )
```

# Data preparation


```python
CLASS_NAMES = {
    0: "background",
    1: "non-neoplastic",
    2: "neoplastic",
}

CLASS_COLORS = {
    0: (0, 0, 0),
    1: (0, 255, 0),
    2: (255, 0, 0),
}


def decode_ground_truth(mask_rgb):
    """
    To RGB mask -> class-index mask 0/1/2.

    Class mapping:
        0 = background      -> black
        1 = non-neoplastic  -> green
        2 = neoplastic      -> red

    Using nearest-color matching to handle or correct any color variations in the mask image.
    This function will convert an RGB mask into a 2D array where each pixel's value corresponds to its class ID (0, 1, or 2).
    """
    mask = np.asarray(mask_rgb)

    if mask.ndim == 2:
        unique_values = np.unique(mask)

        if np.isin(unique_values, [0, 1, 2]).all():
            return mask.astype(np.uint8)

        raise ValueError(
            "Grayscale mask must contain only class IDs 0, 1, 2. "
            f"Found: {unique_values[:20]}"
        )

    if mask.ndim != 3 or mask.shape[-1] != 3:
        raise ValueError(f"Expected HxWx3 RGB mask, got shape={mask.shape}")

    colors = np.array(
        [
            [0, 0, 0],  # background
            [0, 255, 0],  # non-neoplastic
            [255, 0, 0],  # neoplastic
        ],
        dtype=np.float32,
    )

    pixels = mask.astype(np.float32)

    # Shape:
    # pixels -> [H, W, 1, 3]
    # colors -> [1, 1, 3, 3]
    # result -> [H, W, 3]
    diff = pixels[:, :, None, :] - colors[None, None, :, :]

    distances = np.sum(
        diff**2,
        axis=-1,
    )

    class_mask = np.argmin(
        distances,
        axis=-1,
    )

    return class_mask.astype(np.uint8)


def _pick_column(
    df,
    explicit_column=None,
    candidates=None,
    required=True,
):
    if explicit_column is not None:
        if explicit_column not in df.columns:
            raise KeyError(
                f"Column '{explicit_column}' not found. "
                f"Available: {df.columns.tolist()}"
            )

        return explicit_column

    candidates = candidates or []

    normalized = {str(col).strip().lower(): col for col in df.columns}

    for candidate in candidates:
        if candidate.lower() in normalized:
            return normalized[candidate.lower()]

    if len(df.columns) == 1:
        return df.columns[0]

    if required:
        raise KeyError(
            "Cannot infer CSV column automatically. "
            f"Available columns: {df.columns.tolist()}"
        )

    return None


class PolypSegmentationDataset(Dataset):
    IMAGE_COLUMN_CANDIDATES = [
        "image",
        "image_name",
        "image_path",
        "filename",
        "file_name",
        "file",
        "id",
    ]

    MASK_COLUMN_CANDIDATES = [
        "mask",
        "mask_name",
        "mask_path",
        "label",
        "label_path",
        "gt",
        "gt_path",
    ]

    def __init__(
        self,
        dataframe,
        image_dir,
        mask_dir=None,
        transform=None,
        image_col=None,
        mask_col=None,
        has_mask=True,
    ):
        self.df = dataframe.reset_index(drop=True).copy()

        self.image_dir = Path(image_dir)

        self.mask_dir = Path(mask_dir) if mask_dir is not None else None

        self.transform = transform
        self.has_mask = has_mask

        self.image_col = _pick_column(
            self.df,
            explicit_column=image_col,
            candidates=self.IMAGE_COLUMN_CANDIDATES,
            required=True,
        )

        if self.has_mask:
            if self.mask_dir is None:
                raise ValueError("mask_dir is required when has_mask=True.")

            self.mask_col = _pick_column(
                self.df,
                explicit_column=mask_col,
                candidates=self.MASK_COLUMN_CANDIDATES,
                required=True,
            )

            if self.mask_col == self.image_col:
                raise ValueError(
                    "Mask column resolved to the same column as the image column. "
                    "Provide an explicit mask_col or fix the CSV schema."
                )

        else:
            self.mask_col = None

    @staticmethod
    def _resolve_path(root_dir, value):
        candidate = Path(str(value))

        if candidate.is_absolute():
            return candidate

        return Path(root_dir) / str(value)

    def __len__(self):
        return len(self.df)

    def get_image_path(self, index):
        value = self.df.iloc[index][self.image_col]

        return self._resolve_path(
            self.image_dir,
            value,
        )

    def get_mask_path(self, index):
        if not self.has_mask:
            return None

        row = self.df.iloc[index]
        mask_value = row[self.mask_col]

        return self._resolve_path(
            self.mask_dir,
            mask_value,
        )

    def __getitem__(self, index):
        image_path = self.get_image_path(index)

        image_bgr = cv2.imread(
            str(image_path),
            cv2.IMREAD_COLOR,
        )

        if image_bgr is None:
            raise FileNotFoundError(f"Cannot read image: {image_path}")

        image = cv2.cvtColor(
            image_bgr,
            cv2.COLOR_BGR2RGB,
        )

        if not self.has_mask:
            if self.transform is not None:
                transformed = self.transform(image=image)

                image = transformed["image"]

            return {
                "image": image,
                "image_path": str(image_path),
                "image_name": image_path.name,
            }

        mask_path = self.get_mask_path(index)

        mask_raw = cv2.imread(
            str(mask_path),
            cv2.IMREAD_UNCHANGED,
        )

        if mask_raw is None:
            raise FileNotFoundError(f"Cannot read mask: {mask_path}")

        if mask_raw.ndim == 3:
            mask_raw = cv2.cvtColor(
                mask_raw,
                cv2.COLOR_BGR2RGB,
            )

        # Nearest-color RGB -> class ID.
        mask = decode_ground_truth(mask_raw)

        if self.transform is not None:
            transformed = self.transform(
                image=image,
                mask=mask,
            )

            image = transformed["image"]
            mask = transformed["mask"]

        if not torch.is_tensor(mask):
            mask = torch.from_numpy(mask)

        return {
            "image": image,
            "mask": mask.long(),
            "image_path": str(image_path),
            "mask_path": str(mask_path),
            "image_name": image_path.name,
        }
```

```python
class MeanTeacherUnlabeledDataset(Dataset):
    """Return aligned weak and strong views of the same unlabeled image."""

    IMAGE_COLUMN_CANDIDATES = PolypSegmentationDataset.IMAGE_COLUMN_CANDIDATES

    def __init__(
        self,
        dataframe,
        image_dir,
        shared_geometry,
        weak_transform,
        strong_transform,
        image_col=None,
    ):
        self.df = dataframe.reset_index(drop=True).copy()
        self.image_dir = Path(image_dir)
        self.shared_geometry = shared_geometry
        self.weak_transform = weak_transform
        self.strong_transform = strong_transform
        self.image_col = _pick_column(
            self.df,
            explicit_column=image_col,
            candidates=self.IMAGE_COLUMN_CANDIDATES,
            required=True,
        )

    def __len__(self):
        return len(self.df)

    def get_image_path(self, index):
        value = self.df.iloc[index][self.image_col]
        candidate = Path(str(value))
        return candidate if candidate.is_absolute() else self.image_dir / str(value)

    def __getitem__(self, index):
        image_path = self.get_image_path(index)
        image_bgr = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
        if image_bgr is None:
            raise FileNotFoundError(f"Cannot read image: {image_path}")

        image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

        # Geometry is sampled exactly once. Only photometric appearance differs.
        aligned_image = self.shared_geometry(image=image)["image"]
        weak_image = self.weak_transform(image=aligned_image)["image"]
        strong_image = self.strong_transform(image=aligned_image)["image"]

        return {
            "weak_image": weak_image,
            "strong_image": strong_image,
            "image_path": str(image_path),
            "image_name": image_path.name,
        }


class MeanTeacherLabeledDataset(Dataset):
    """Return aligned weak (Teacher) and strong (Student) views of the same
    labeled image, plus the mask aligned to the shared geometry.

    This mirrors MeanTeacherUnlabeledDataset so labeled examples also
    receive a consistency signal, matching the original Mean Teacher design
    where the consistency cost is evaluated on all input data, not only the
    unlabeled subset.
    """

    IMAGE_COLUMN_CANDIDATES = PolypSegmentationDataset.IMAGE_COLUMN_CANDIDATES
    MASK_COLUMN_CANDIDATES = PolypSegmentationDataset.MASK_COLUMN_CANDIDATES

    def __init__(
        self,
        dataframe,
        image_dir,
        mask_dir,
        shared_geometry,
        weak_transform,
        strong_transform,
        image_col=None,
        mask_col=None,
    ):
        self.df = dataframe.reset_index(drop=True).copy()
        self.image_dir = Path(image_dir)
        self.mask_dir = Path(mask_dir)
        self.shared_geometry = shared_geometry
        self.weak_transform = weak_transform
        self.strong_transform = strong_transform
        self.image_col = _pick_column(
            self.df,
            explicit_column=image_col,
            candidates=self.IMAGE_COLUMN_CANDIDATES,
            required=True,
        )
        self.mask_col = _pick_column(
            self.df,
            explicit_column=mask_col,
            candidates=self.MASK_COLUMN_CANDIDATES,
            required=True,
        )

        if self.mask_col == self.image_col:
            raise ValueError(
                "Mask column resolved to the same column as the image column. "
                "Provide an explicit mask_col or fix the CSV schema."
            )

    def __len__(self):
        return len(self.df)

    def get_image_path(self, index):
        value = self.df.iloc[index][self.image_col]
        candidate = Path(str(value))
        return candidate if candidate.is_absolute() else self.image_dir / str(value)

    def get_mask_path(self, index):
        row = self.df.iloc[index]
        mask_value = row[self.mask_col]
        candidate = Path(str(mask_value))
        return candidate if candidate.is_absolute() else self.mask_dir / str(mask_value)

    def __getitem__(self, index):
        image_path = self.get_image_path(index)
        image_bgr = cv2.imread(str(image_path), cv2.IMREAD_COLOR)
        if image_bgr is None:
            raise FileNotFoundError(f"Cannot read image: {image_path}")
        image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)

        mask_path = self.get_mask_path(index)
        mask_raw = cv2.imread(str(mask_path), cv2.IMREAD_UNCHANGED)
        if mask_raw is None:
            raise FileNotFoundError(f"Cannot read mask: {mask_path}")
        if mask_raw.ndim == 3:
            mask_raw = cv2.cvtColor(mask_raw, cv2.COLOR_BGR2RGB)
        mask = decode_ground_truth(mask_raw)

        # Geometry sampled once so mask, weak view and strong view stay aligned.
        aligned = self.shared_geometry(image=image, mask=mask)
        aligned_image, aligned_mask = aligned["image"], aligned["mask"]

        weak_image = self.weak_transform(image=aligned_image)["image"]
        strong_image = self.strong_transform(image=aligned_image)["image"]

        if not torch.is_tensor(aligned_mask):
            aligned_mask = torch.from_numpy(aligned_mask)

        return {
            "weak_image": weak_image,
            "strong_image": strong_image,
            "mask": aligned_mask.long(),
            "image_path": str(image_path),
            "mask_path": str(mask_path),
            "image_name": image_path.name,
        }


def validate_labeled_dataset(dataset, dataset_name="dataset", max_samples=8):
    """Fail early if image/mask columns or representative mask paths are wrong."""
    if getattr(dataset, "mask_col", None) is None:
        raise ValueError(f"[{dataset_name}] mask_col was not resolved.")

    if dataset.image_col == dataset.mask_col:
        raise ValueError(
            f"[{dataset_name}] image_col and mask_col are both '{dataset.image_col}'."
        )

    if len(dataset) == 0:
        raise ValueError(f"[{dataset_name}] dataset is empty.")

    sample_count = min(int(max_samples), len(dataset))
    indices = np.unique(np.linspace(0, len(dataset) - 1, num=sample_count, dtype=int))
    seen_classes = set()

    for index in indices:
        image_path = Path(dataset.get_image_path(int(index)))
        mask_path = Path(dataset.get_mask_path(int(index)))

        if not image_path.exists():
            raise FileNotFoundError(f"[{dataset_name}] Missing image: {image_path}")
        if not mask_path.exists():
            raise FileNotFoundError(f"[{dataset_name}] Missing mask: {mask_path}")

        if image_path.resolve() == mask_path.resolve():
            raise ValueError(
                f"[{dataset_name}] Image and mask resolve to the same file: {image_path}"
            )

        mask_raw = cv2.imread(str(mask_path), cv2.IMREAD_UNCHANGED)
        if mask_raw is None:
            raise FileNotFoundError(f"[{dataset_name}] Cannot read mask: {mask_path}")
        if mask_raw.ndim == 3:
            mask_raw = cv2.cvtColor(mask_raw, cv2.COLOR_BGR2RGB)

        decoded = decode_ground_truth(mask_raw)
        unique_classes = set(np.unique(decoded).tolist())
        if not unique_classes.issubset({0, 1, 2}):
            raise ValueError(
                f"[{dataset_name}] Invalid decoded class IDs {sorted(unique_classes)} "
                f"in mask {mask_path}"
            )
        seen_classes.update(unique_classes)

    print(
        f"[Mask check] {dataset_name}: image_col='{dataset.image_col}', "
        f"mask_col='{dataset.mask_col}', checked={len(indices)}, "
        f"classes_seen={sorted(seen_classes)}"
    )
```

# Architecture


## U-Net


```python
class ResNetEncoder(nn.Module):
    """
    ResNet encoder backbone.

    Supports:
        resnet34, resnet50

    Returns:
        [x0, x1, x2, x3, x4]
    """

    def __init__(
        self,
        backbone="resnet34",
        pretrained=True,
    ):
        super().__init__()

        backbone = backbone.lower()

        if backbone == "resnet34":
            weights = ResNet34_Weights.DEFAULT if pretrained else None
            model = resnet34(weights=weights)
            self.out_channels = [64, 64, 128, 256, 512]

        elif backbone == "resnet50":
            weights = ResNet50_Weights.DEFAULT if pretrained else None
            model = resnet50(weights=weights)
            self.out_channels = [64, 256, 512, 1024, 2048]

        else:
            raise ValueError(
                f"Unsupported backbone: {backbone}. "
                "Supported: resnet18, resnet34, "
                "resnet50, resnet101"
            )

        self.conv1 = model.conv1
        self.bn1 = model.bn1
        self.relu = model.relu
        self.maxpool = model.maxpool

        self.layer1 = model.layer1
        self.layer2 = model.layer2
        self.layer3 = model.layer3
        self.layer4 = model.layer4

    def forward(self, x):
        x0 = self.relu(self.bn1(self.conv1(x)))  # H/2

        x = self.maxpool(x0)  # H/4
        x1 = self.layer1(x)  # H/4
        x2 = self.layer2(x1)  # H/8
        x3 = self.layer3(x2)  # H/16
        x4 = self.layer4(x3)  # H/32

        return [x0, x1, x2, x3, x4]

    def freeze(self):
        for parameter in self.parameters():
            parameter.requires_grad = False

    def unfreeze(self):
        for parameter in self.parameters():
            parameter.requires_grad = True
```

```python
class DoubleConv(nn.Module):
    """
    (Conv2d -> BatchNorm -> ReLU) * 2
    """

    def __init__(
        self,
        in_channels,
        out_channels,
    ):
        super().__init__()

        self.double_conv = nn.Sequential(
            nn.Conv2d(
                in_channels,
                out_channels,
                kernel_size=3,
                padding=1,
                bias=False,
            ),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
            nn.Conv2d(
                out_channels,
                out_channels,
                kernel_size=3,
                padding=1,
                bias=False,
            ),
            nn.BatchNorm2d(out_channels),
            nn.ReLU(inplace=True),
        )

    def forward(self, x):
        return self.double_conv(x)


class Up(nn.Module):
    """
    Upsample decoder feature, concat encoder skip, before DoubleConv.
    """

    def __init__(
        self,
        decoder_channels,
        skip_channels,
        out_channels,
        bilinear=False,
    ):
        super().__init__()

        if bilinear:
            self.up = nn.Sequential(
                nn.Upsample(
                    scale_factor=2,
                    mode="bilinear",
                    align_corners=True,
                ),
                nn.Conv2d(
                    decoder_channels,
                    out_channels,
                    kernel_size=1,
                ),
            )
        else:
            self.up = nn.ConvTranspose2d(
                decoder_channels,
                out_channels,
                kernel_size=2,
                stride=2,
            )

        self.conv = DoubleConv(
            out_channels + skip_channels,
            out_channels,
        )

    def forward(self, x, skip):
        x = self.up(x)

        diff_y = skip.size(2) - x.size(2)
        diff_x = skip.size(3) - x.size(3)

        if diff_x != 0 or diff_y != 0:
            x = F.pad(
                x,
                [
                    diff_x // 2,
                    diff_x - diff_x // 2,
                    diff_y // 2,
                    diff_y - diff_y // 2,
                ],
            )

        x = torch.cat([skip, x], dim=1)

        return self.conv(x)


class OutConv(nn.Module):
    def __init__(
        self,
        in_channels,
        n_classes,
    ):
        super().__init__()

        self.conv = nn.Conv2d(
            in_channels,
            n_classes,
            kernel_size=1,
        )

    def forward(self, x):
        return self.conv(x)


class UNet(nn.Module):
    """
    U-Net with ResNet encoder backbone.

    Output:
        logits: (B, n_classes, H, W)

    return_features=True:
        return encoder/bottleneck/decoder features.
    """

    def __init__(
        self,
        backbone="resnet34",
        n_classes=3,
        pretrained=True,
        bilinear=False,
    ):
        super().__init__()

        self.backbone_name = backbone
        self.n_classes = n_classes
        self.pretrained = pretrained
        self.bilinear = bilinear

        self.encoder = ResNetEncoder(
            backbone=backbone,
            pretrained=pretrained,
        )

        c0, c1, c2, c3, c4 = self.encoder.out_channels

        self.up4 = Up(
            decoder_channels=c4,
            skip_channels=c3,
            out_channels=256,
            bilinear=bilinear,
        )

        self.up3 = Up(
            decoder_channels=256,
            skip_channels=c2,
            out_channels=128,
            bilinear=bilinear,
        )

        self.up2 = Up(
            decoder_channels=128,
            skip_channels=c1,
            out_channels=64,
            bilinear=bilinear,
        )

        self.up1 = Up(
            decoder_channels=64,
            skip_channels=c0,
            out_channels=64,
            bilinear=bilinear,
        )

        if bilinear:
            self.final_up = nn.Sequential(
                nn.Upsample(
                    scale_factor=2,
                    mode="bilinear",
                    align_corners=True,
                ),
                nn.Conv2d(
                    64,
                    32,
                    kernel_size=1,
                ),
                nn.BatchNorm2d(32),
                nn.ReLU(inplace=True),
            )
        else:
            self.final_up = nn.Sequential(
                nn.ConvTranspose2d(
                    64,
                    32,
                    kernel_size=2,
                    stride=2,
                ),
                nn.BatchNorm2d(32),
                nn.ReLU(inplace=True),
            )

        self.outc = OutConv(
            32,
            n_classes,
        )

    def forward(
        self,
        x,
        return_features=False,
    ):
        input_size = x.shape[-2:]

        features = self.encoder(x)
        x0, x1, x2, x3, x4 = features

        x = self.up4(x4, x3)
        x = self.up3(x, x2)
        x = self.up2(x, x1)
        x = self.up1(x, x0)

        decoder_feature = self.final_up(x)

        logits = self.outc(decoder_feature)

        if logits.shape[-2:] != input_size:
            logits = F.interpolate(
                logits,
                size=input_size,
                mode="bilinear",
                align_corners=False,
            )

        if return_features:
            return logits, {
                "encoder_0": x0,
                "encoder_1": x1,
                "encoder_2": x2,
                "encoder_3": x3,
                "bottleneck": x4,
                "decoder": decoder_feature,
            }

        return logits

    def freeze_encoder(self):
        self.encoder.freeze()

    def unfreeze_encoder(self):
        self.encoder.unfreeze()

class ContrastiveUNet(UNet):
    """Normal U-Net inference plus a Student-only patch embedding path."""

    def __init__(self, contrastive_config, **kwargs):
        super().__init__(**kwargs)
        self.contrastive_config = copy.deepcopy(contrastive_config)
        dim = int(contrastive_config["embedding_dim"])
        # Do not consume the global RNG stream used by data/model training.
        with torch.random.fork_rng(devices=[]):
            torch.manual_seed(int(contrastive_config["seed"]))
            self.projection_head = nn.Sequential(
                nn.Linear(self.encoder.out_channels[-1], dim),
                nn.ReLU(inplace=False),
                nn.Linear(dim, dim),
            )

    def embed_patches(self, patches):
        # Small attended patches have different statistics from full images.
        # Freeze only BN statistics here, not encoder weights or gradients.
        bn_layers = [m for m in self.encoder.modules()
                     if isinstance(m, nn.modules.batchnorm._BatchNorm)]
        modes = [m.training for m in bn_layers]
        try:
            for m in bn_layers:
                m.eval()
            features = self.encoder(patches)[-1]
            pooled = F.adaptive_avg_pool2d(features, 1).flatten(1)
            return F.normalize(self.projection_head(pooled).float(), dim=1)
        finally:
            for m, mode in zip(bn_layers, modes):
                m.train(mode)
```

# Metrics and losses


## metrics


```python
def _confusion_counts_from_logits(
    logits,
    targets,
    num_classes=3,
):
    """
    Aggregate TP/FP/FN over all pixels in the supplied batch.
    Counts are returned per class and can be accumulated across batches.
    """
    preds = torch.argmax(logits, dim=1)

    tp = np.zeros(num_classes, dtype=np.float64)
    fp = np.zeros(num_classes, dtype=np.float64)
    fn = np.zeros(num_classes, dtype=np.float64)

    for class_id in range(num_classes):
        pred_c = preds == class_id
        true_c = targets == class_id

        tp[class_id] = (pred_c & true_c).sum().item()
        fp[class_id] = (pred_c & ~true_c).sum().item()
        fn[class_id] = (~pred_c & true_c).sum().item()

    return tp, fp, fn


def _macro_dice_from_counts(
    tp,
    fp,
    fn,
    include_background=False,
):
    """
    Macro Dice from dataset/epoch-level TP/FP/FN counts.

    A class whose denominator is zero over the whole evaluated set is skipped
    from the macro average, matching the existing notebook behavior for an
    entirely absent class without letting batch composition bias the score.
    """
    start_class = 0 if include_background else 1
    scores = []

    for class_id in range(start_class, len(tp)):
        denominator = 2.0 * tp[class_id] + fp[class_id] + fn[class_id]
        if denominator > 0:
            scores.append((2.0 * tp[class_id]) / denominator)

    return float(np.mean(scores)) if scores else 0.0


def multiclass_dice_score(
    logits,
    targets,
    num_classes=3,
    include_background=False,
):
    """
    Macro Dice computed from TP/FP/FN aggregated over the supplied batch.

    Training/validation functions accumulate these counts across the full
    epoch before reporting Dice, so the final score is independent of how the
    dataset happens to be split into batches.
    """
    tp, fp, fn = _confusion_counts_from_logits(
        logits,
        targets,
        num_classes=num_classes,
    )
    score = _macro_dice_from_counts(
        tp,
        fp,
        fn,
        include_background=include_background,
    )
    return torch.tensor(score, dtype=torch.float32, device=logits.device)


def hd95_binary(pred_mask, true_mask):
    """
    Directional-max symmetric HD95 for one binary mask pair.

    The 95th percentile is computed separately in both directions and the
    larger value is returned, matching the non-directed MONAI/DeepMind style.

    Empty-mask convention:
        - both prediction and ground truth empty -> NaN (excluded from mean)
        - exactly one side empty                 -> image-diagonal penalty
    """
    pred = np.asarray(pred_mask, dtype=bool)
    true = np.asarray(true_mask, dtype=bool)

    pred_empty = not pred.any()
    true_empty = not true.any()

    if pred_empty and true_empty:
        return np.nan

    if pred_empty or true_empty:
        height, width = true.shape[-2:]
        return float(np.hypot(height, width))

    pred_surface = pred ^ binary_erosion(pred)
    true_surface = true ^ binary_erosion(true)

    if not pred_surface.any():
        pred_surface = pred
    if not true_surface.any():
        true_surface = true

    distance_to_true = distance_transform_edt(~true_surface)
    distance_to_pred = distance_transform_edt(~pred_surface)

    distances_pred_to_true = distance_to_true[pred_surface]
    distances_true_to_pred = distance_to_pred[true_surface]

    directed_hd95_pred_to_true = np.percentile(
        distances_pred_to_true,
        95,
    )
    directed_hd95_true_to_pred = np.percentile(
        distances_true_to_pred,
        95,
    )

    return float(
        max(
            directed_hd95_pred_to_true,
            directed_hd95_true_to_pred,
        )
    )


def _safe_divide(numerator, denominator):
    if denominator == 0:
        return np.nan
    return float(numerator / denominator)
```

## losses


```python
class GeneralizedDiceLoss(nn.Module):
    """Generalized Dice Loss (GDL): Automatically balances class weights based on the inverse square of the true area."""

    def __init__(self, num_classes=3, smooth=1e-6, ignore_index=None):
        super().__init__()
        self.num_classes = num_classes
        self.smooth = smooth
        self.ignore_index = ignore_index

    def forward(self, logits, targets):
        probabilities = torch.softmax(logits, dim=1)

        if self.ignore_index is not None:
            valid_mask = targets != self.ignore_index
            safe_targets = torch.where(valid_mask, targets, torch.zeros_like(targets))
        else:
            valid_mask = torch.ones_like(targets, dtype=torch.bool)
            safe_targets = targets

        targets_one_hot = F.one_hot(safe_targets.long(), num_classes=self.num_classes)
        targets_one_hot = targets_one_hot.permute(0, 3, 1, 2).float()
        valid_mask = valid_mask.unsqueeze(1).float()
        probabilities = probabilities * valid_mask
        targets_one_hot = targets_one_hot * valid_mask

        dims = (0, 2, 3)
        targets_sum = torch.sum(targets_one_hot, dim=dims)

        present = targets_sum > 0
        safe_targets_sum = torch.where(
            present,
            targets_sum,
            torch.ones_like(targets_sum),
        )
        weights = present.to(probabilities.dtype) / (
            safe_targets_sum + self.smooth
        ).pow(2)
        weights = weights / weights.sum().clamp_min(self.smooth)

        intersection = torch.sum(probabilities * targets_one_hot, dim=dims)
        pred_sum = torch.sum(probabilities, dim=dims)

        weighted_intersection = torch.sum(weights * intersection)
        weighted_denominator = torch.sum(weights * (pred_sum + targets_sum))

        gdl = (2.0 * weighted_intersection + self.smooth) / (
            weighted_denominator + self.smooth
        )
        return 1.0 - gdl


class WeightedCE_GDL_Loss(nn.Module):
    def __init__(
        self,
        class_weights=None,
        num_classes=3,
        ce_loss_weight=1.0,
        gdl_loss_weight=1.0,
        smooth=1e-6,
        ignore_index=-100,
    ):
        super().__init__()
        self.ce_loss_weight = float(ce_loss_weight)
        self.gdl_loss_weight = float(gdl_loss_weight)

        ce_weights = (
            torch.tensor(class_weights, dtype=torch.float32)
            if class_weights is not None
            else None
        )

        self.register_buffer("ce_weights", ce_weights, persistent=True)

        self.cross_entropy = nn.CrossEntropyLoss(
            weight=self.ce_weights, ignore_index=ignore_index
        )
        self.gdl = GeneralizedDiceLoss(
            num_classes=num_classes, smooth=smooth, ignore_index=ignore_index
        )

    def get_extra_state(self):
        return {
            "ce_loss_weight": self.ce_loss_weight,
            "gdl_loss_weight": self.gdl_loss_weight,
            "num_classes": self.gdl.num_classes,
            "smooth": self.gdl.smooth,
            "ignore_index": self.gdl.ignore_index,
        }

    def set_extra_state(self, state):
        if not state:
            return

        self.ce_loss_weight = float(state["ce_loss_weight"])
        self.gdl_loss_weight = float(state["gdl_loss_weight"])
        self.gdl.num_classes = int(state["num_classes"])
        self.gdl.smooth = float(state["smooth"])
        self.gdl.ignore_index = state["ignore_index"]
        self.cross_entropy.ignore_index = state["ignore_index"]

    def forward(self, logits, targets):
        ce_loss = self.cross_entropy(logits, targets)
        gdl_loss = self.gdl(logits, targets)
        return self.ce_loss_weight * ce_loss + self.gdl_loss_weight * gdl_loss
```

## Mean Teacher consistency loss

This notebook uses the authors' default **softmax-MSE consistency loss**:

- Paper: https://arxiv.org/abs/1703.01780
- Official loss implementation: https://github.com/CuriousAI/mean-teacher/blob/master/pytorch/mean_teacher/losses.py
- Official sigmoid ramp-up and EMA update: https://github.com/CuriousAI/mean-teacher

The original implementation was written for classification tensors `(B, C)`. Here, `reduction="mean"` applies the same normalization principle over segmentation tensors `(B, C, H, W)`. Gradients flow only through Student predictions; Teacher predictions are detached.


```python
class SoftmaxMSEConsistencyLoss(nn.Module):
    """
    Mean Teacher softmax-MSE consistency loss.

    Source:
        CuriousAI/mean-teacher, pytorch/mean_teacher/losses.py

    Segmentation adaptation:
        Mean reduction over B, C, H and W. This is equivalent to the
        authors' summed loss divided by batch size and spatial positions.
    """

    def forward(self, student_logits, teacher_logits):
        if student_logits.shape != teacher_logits.shape:
            raise ValueError(
                "Student and Teacher logits must have identical shapes, "
                f"got {student_logits.shape} and {teacher_logits.shape}."
            )

        student_probabilities = torch.softmax(student_logits, dim=1)
        teacher_probabilities = torch.softmax(
            teacher_logits.detach(),
            dim=1,
        )

        return F.mse_loss(
            student_probabilities,
            teacher_probabilities,
            reduction="mean",
        )

CONTRASTIVE_METRIC_NAMES = (
    "contrastive_loss", "weighted_contrastive_loss", "contrastive_weight",
    "contrastive_to_supervised_ratio", "contrastive_active_fraction",
    "contrastive_anchors", "contrastive_patches", "pseudo_coverage",
    "covariance_scale", "patches_background", "patches_non_neoplastic",
    "patches_neoplastic",
)


def validate_contrastive_config(cfg):
    if cfg["objective"] not in {"infonce", "plgcl"}:
        raise ValueError("objective must be infonce or plgcl")
    if not np.isfinite(cfg["temperature"]) or cfg["temperature"] <= 0:
        raise ValueError("temperature must be finite and positive")
    for key in ("max_weight", "covariance_scale"):
        if not np.isfinite(cfg[key]) or cfg[key] < 0:
            raise ValueError(f"{key} must be finite and nonnegative")
    for key in ("warmup_epochs", "rampup_epochs"):
        if int(cfg[key]) != cfg[key] or cfg[key] < 0:
            raise ValueError(f"{key} must be a nonnegative integer")
    for key in ("embedding_dim", "patch_size", "max_patches_per_class",
                "anchors_per_class", "positives_per_anchor", "negative_keys_per_class", "min_keys"):
        if int(cfg[key]) != cfg[key] or cfg[key] < 1:
            raise ValueError(f"{key} must be a positive integer")
    if cfg["min_keys"] < 2:
        raise ValueError("min_keys must be >= 2 for covariance estimates")
    if cfg["max_patches_per_class"] < cfg["min_keys"] + 1:
        raise ValueError("Need an anchor plus at least min_keys same-class patches")
    if min(cfg["positives_per_anchor"], cfg["negative_keys_per_class"]) < cfg["min_keys"]:
        raise ValueError("Key limits must be >= min_keys")
    for key in ("pseudo_threshold", "min_valid_fraction", "min_class_fraction"):
        if not 0 < cfg[key] <= 1:
            raise ValueError(f"{key} must be in (0, 1]")
    if cfg["min_class_fraction"] <= 0.5:
        raise ValueError("Require a majority class to avoid duplicate region labels")


def contrastive_schedule(epoch, cfg):
    elapsed = int(epoch) - int(cfg["warmup_epochs"])
    ramp = int(cfg["rampup_epochs"])
    factor = 0.0 if elapsed <= 0 else (1.0 if ramp == 0 else min(1.0, elapsed / ramp))
    return cfg["max_weight"] * factor, cfg["covariance_scale"] * factor


@torch.no_grad()
def sample_student_patches(images, student_logits, labeled_masks, num_labeled, cfg, step):
    """Bounded RGB attended-patch sampler, shared verbatim by both experiments.

    Input images use the notebook's ImageNet normalization. Sampling is detached;
    pseudo-label selection must not become a differentiable shortcut for the loss.
    Local RNG leaves loader/augmentation RNG streams unchanged.
    """
    batch, channels, height, width = images.shape
    size = int(cfg["patch_size"])
    if channels != 3 or height < size or width < size:
        raise ValueError("RGB images must be at least patch_size in both dimensions")
    probabilities = torch.softmax(student_logits.detach().float(), dim=1)
    if probabilities.shape[-2:] != (height, width):
        raise ValueError("Patch selection requires aligned full-resolution predictions")
    classes = probabilities.shape[1]
    confidence, labels = probabilities.max(dim=1)
    valid = confidence >= cfg["pseudo_threshold"]
    if num_labeled:
        gt_valid = (labeled_masks >= 0) & (labeled_masks < classes)
        safe_gt = labeled_masks.clamp(0, classes - 1)
        probabilities[:num_labeled] = F.one_hot(safe_gt, classes).permute(0, 3, 1, 2).float()
        probabilities[:num_labeled] *= gt_valid.unsqueeze(1)
        labels[:num_labeled] = safe_gt
        valid[:num_labeled] = gt_valid
    coverage = float(valid[num_labeled:].float().mean()) if batch > num_labeled else 0.0
    mean = images.new_tensor([0.485, 0.456, 0.406])[None, :, None, None]
    std = images.new_tensor([0.229, 0.224, 0.225])[None, :, None, None]
    rgb = (images.detach() * std + mean).clamp(0, 1)
    # Include the right/bottom edge even when the input is not a multiple of size.
    ys = sorted(set(range(0, height - size + 1, size)) | {height - size})
    xs = sorted(set(range(0, width - size + 1, size)) | {width - size})
    # Pool eligibility on-device, transfer the small grid once rather than sync per patch.
    candidates = [[] for _ in range(classes)]
    masks = torch.stack([valid] + [(labels == c) & valid for c in range(classes)], dim=1).float()
    grid = torch.stack([masks[:, :, y:y+size, x:x+size].mean(dim=(2, 3))
                        for y in ys for x in xs], dim=1).cpu().numpy()
    locations = [(y, x) for y in ys for x in xs]
    for b in range(batch):
        for pos, (y, x) in enumerate(locations):
            if grid[b, pos, 0] < cfg["min_valid_fraction"]:
                continue
            for c in range(classes):
                if grid[b, pos, c + 1] >= cfg["min_class_fraction"]:
                    candidates[c].append((b, y, x))
    rng = np.random.default_rng(int(cfg["seed"]) + int(step))
    patches, patch_labels, entropies = [], [], []
    for c, options in enumerate(candidates):
        take = min(len(options), int(cfg["max_patches_per_class"]))
        indices = rng.choice(len(options), size=take, replace=False) if take else []
        for index in indices:
            b, y, x = options[index]
            attended = (rgb[b:b+1, :, y:y+size, x:x+size]
                        * probabilities[b:b+1, c:c+1, y:y+size, x:x+size]
                        * valid[b:b+1, None, y:y+size, x:x+size])
            # Eq. (2)-(3) inspired RGB extension: mean over channels and pixels.
            p = attended.clamp(1e-6, 1.0 - 1e-6)
            entropies.append((-(p * p.log() + (1-p) * (1-p).log())).mean())
            patches.append((attended - mean) / std)
            patch_labels.append(c)
    if not patches:
        return (images.new_empty((0, channels, size, size)),
                torch.empty(0, device=images.device, dtype=torch.long),
                images.new_empty(0), coverage)
    return (torch.cat(patches), torch.tensor(patch_labels, device=images.device),
            torch.stack(entropies), coverage)


def make_contrastive_groups(labels, entropies, cfg, step):
    """Same query/positive/negative indices for InfoNCE and PLGCL.

    Ignore non-selected same-class patches; never use them as negatives.
    Positive keys exclude the anchor. All present other classes contribute equal
    key counts, so zeta in PLGCL is exactly that count, not an arbitrary multiplier.
    """
    y = labels.detach().cpu().numpy()
    entropy = entropies.detach().float().cpu().numpy()
    rng = np.random.default_rng(int(cfg["seed"]) + int(step) + 100000)
    pools = {int(c): np.flatnonzero(y == c) for c in np.unique(y)}
    groups = []
    for c, pool in pools.items():
        other = {k: v for k, v in pools.items() if k != c and len(v) >= cfg["min_keys"]}
        if len(pool) <= cfg["min_keys"] or not other:
            continue
        count = min(int(cfg["negative_keys_per_class"]), min(map(len, other.values())))
        for anchor in rng.choice(pool, size=min(len(pool), cfg["anchors_per_class"]), replace=False):
            positives = pool[pool != anchor]
            order = np.argsort(np.abs(entropy[positives] - entropy[anchor]), kind="stable")
            positives = positives[order[:cfg["positives_per_anchor"]]]
            negatives = {k: rng.choice(v, size=count, replace=False).tolist() for k, v in other.items()}
            groups.append((c, int(anchor), positives.tolist(), negatives))
    return groups


def student_contrastive_loss(embeddings, groups, objective, temperature, covariance_scale):
    """Class-balanced empirical InfoNCE or PLGCL Gaussian upper-bound objective.

    For a query q and a key group K, var(K @ q, unbiased=False) is exactly
    q^T Cov_population(K) q, including off-diagonal covariance terms. No diagonal
    approximation and no detached prototypes: all sampled Student embeddings learn.
    """
    if objective not in {"infonce", "plgcl"}:
        raise ValueError("Unknown contrastive objective")
    if temperature <= 0 or covariance_scale < 0:
        raise ValueError("Invalid contrastive temperature/covariance scale")
    if not groups:
        return embeddings.sum() * 0.0
    z = F.normalize(embeddings.float(), dim=1)
    class_losses = {}
    for c, anchor, positive_ids, negatives_by_class in groups:
        q = z[anchor]
        positive_dot = z[positive_ids] @ q
        negative_dot = [z[ids] @ q for ids in negatives_by_class.values()]
        if objective == "infonce":
            # Average single-positive InfoNCE over chosen positive keys (Eq. 4 style).
            # Each denominator contains that positive plus the SAME negative pool.
            pos_logits = positive_dot / temperature
            neg_logsum = torch.logsumexp(torch.cat(negative_dot) / temperature, dim=0)
            loss = (torch.logaddexp(pos_logits, neg_logsum) - pos_logits).mean()
        else:
            positive_mean = positive_dot.mean() / temperature
            scores = [positive_mean + covariance_scale * positive_dot.var(unbiased=False)
                      / (2 * temperature**2)]
            for dot in negative_dot:
                scores.append(dot.mean() / temperature
                              + covariance_scale * dot.var(unbiased=False) / (2 * temperature**2)
                              + np.log(dot.numel()))  # log zeta: actual negative count
            loss = torch.logsumexp(torch.stack(scores), dim=0) - positive_mean
        class_losses.setdefault(c, []).append(loss)
    return torch.stack([torch.stack(values).mean() for values in class_losses.values()]).mean()


def compute_student_contrastive(student, images, student_logits, labeled_masks,
                                num_labeled, epoch, step):
    cfg = student.contrastive_config
    weight, covariance_scale = contrastive_schedule(epoch, cfg)
    metrics = {name: 0.0 for name in CONTRASTIVE_METRIC_NAMES}
    metrics.update(contrastive_weight=weight, covariance_scale=covariance_scale)
    zero = student_logits.sum() * 0.0
    if weight == 0:
        return zero, metrics
    patches, labels, entropy, coverage = sample_student_patches(
        images, student_logits, labeled_masks, num_labeled, cfg, step)
    metrics["pseudo_coverage"] = coverage
    metrics["contrastive_patches"] = float(len(labels))
    for c, name in enumerate(("patches_background", "patches_non_neoplastic", "patches_neoplastic")):
        metrics[name] = float((labels == c).sum())
    groups = make_contrastive_groups(labels, entropy, cfg, step)
    if not groups:
        return zero, metrics
    embeddings = student.embed_patches(patches)
    loss = student_contrastive_loss(embeddings, groups, cfg["objective"],
                                    cfg["temperature"], covariance_scale)
    if not torch.isfinite(loss):
        raise FloatingPointError("Non-finite contrastive loss; inspect inputs and settings")
    metrics.update(contrastive_loss=float(loss.detach()),
                   weighted_contrastive_loss=float(loss.detach()) * weight,
                   contrastive_active_fraction=1.0, contrastive_anchors=float(len(groups)))
    return weight * loss, metrics
```

# Mean Teacher utils


```python
def create_teacher(student):
    """Create the EMA Teacher.

    Teacher stays in train mode during training so its BatchNorm layers use
    batch statistics just like the Student. Gradients remain disabled.
    """
    teacher = copy.deepcopy(student)
    teacher.requires_grad_(False)
    teacher.train()
    return teacher


@torch.no_grad()
def update_ema_teacher(student, teacher, ema_decay, global_step):
    """EMA-update Teacher parameters only (official Mean Teacher behavior).

    BatchNorm buffers are NOT copied from Student. While Teacher is in train
    mode, its own running statistics are updated by its weak-view forward pass.
    """
    effective_decay = min(
        float(ema_decay),
        1.0 - 1.0 / (float(global_step) + 1.0),
    )

    student_parameters = dict(student.named_parameters())
    teacher_parameters = dict(teacher.named_parameters())

    if student_parameters.keys() != teacher_parameters.keys():
        raise RuntimeError("Student and Teacher parameter structures do not match.")

    for name, teacher_parameter in teacher_parameters.items():
        teacher_parameter.mul_(effective_decay).add_(
            student_parameters[name],
            alpha=1.0 - effective_decay,
        )


def sigmoid_rampup(current, rampup_length):
    """Official Mean Teacher exponential sigmoid ramp-up."""
    rampup_length = float(rampup_length)
    if rampup_length == 0.0:
        return 1.0

    current = float(np.clip(current, 0.0, rampup_length))
    phase = 1.0 - current / rampup_length
    return float(np.exp(-5.0 * phase * phase))


def get_consistency_weight(epoch, mt_config):
    maximum = float(mt_config["max_consistency_weight"])
    rampup = int(mt_config["consistency_rampup_epochs"])
    return maximum * sigmoid_rampup(
        current=max(epoch - 1, 0),
        rampup_length=rampup,
    )
```

# Training and checkpoint utils


```python
def clone_state_dict_to_cpu(model):
    return {
        key: value.detach().cpu().clone() for key, value in model.state_dict().items()
    }


def component_type(component):
    if component is None:
        return None
    cls = component.__class__
    return f"{cls.__module__}.{cls.__qualname__}"


def empty_mean_teacher_history(num_classes=None):
    return {
        **{name: [] for name in CONTRASTIVE_METRIC_NAMES},
        "supervised_loss": [],
        "consistency_loss": [],
        "weighted_consistency_loss": [],
        "consistency_to_supervised_ratio": [],
        "total_loss": [],
        "train_dice": [],
        "student_val_loss": [],
        "student_val_dice": [],
        "teacher_val_loss": [],
        "teacher_val_dice": [],
        "consistency_weight": [],
        "learning_rate": [],
    }


def _next_cycled_batch(loader, iterator):
    try:
        batch = next(iterator)
    except StopIteration:
        iterator = iter(loader)
        batch = next(iterator)
    return batch, iterator


def validate_resume_component(checkpoint, key, component):
    saved_type = checkpoint.get(f"{key}_type")
    current_type = component_type(component)
    if saved_type != current_type:
        raise ValueError(
            f"Cannot resume {key}: checkpoint uses {saved_type}, "
            f"but the notebook constructs {current_type}."
        )
```

```python
def save_mean_teacher_checkpoint(
    student,
    teacher,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    scheduler,
    epoch,
    global_step,
    best_loss,
    best_dice,
    best_epoch,
    best_student_state_dict,
    best_teacher_state_dict,
    early_stopping_counter,
    history,
    config,
    mean_teacher_config,
    checkpoint_dir,
):
    checkpoint_dir = Path(checkpoint_dir)
    checkpoint_dir.mkdir(parents=True, exist_ok=True)
    timestamp = datetime.now().strftime(config.get("timestamp_format", "%y%m%d_%H%M%S"))
    path = (
        checkpoint_dir
        / f"{EXPERIMENT_NAME}_{student.__class__.__name__}_epoch_{epoch:03d}_{timestamp}.pth"
    )

    payload = {
        "training_method": EXPERIMENT_NAME,
        "epoch": int(epoch),
        "global_step": int(global_step),
        "student_type": component_type(student),
        "student_state_dict": student.state_dict(),
        "teacher_type": component_type(teacher),
        "teacher_state_dict": teacher.state_dict(),
        "supervised_criterion_type": component_type(supervised_criterion),
        "supervised_criterion_state_dict": supervised_criterion.state_dict(),
        "consistency_criterion_type": component_type(consistency_criterion),
        "consistency_criterion_state_dict": consistency_criterion.state_dict(),
        "optimizer_type": component_type(optimizer),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_type": component_type(scheduler),
        "scheduler_state_dict": (
            scheduler.state_dict() if scheduler is not None else None
        ),
        "best_loss": float(best_loss),
        "best_dice": float(best_dice),
        "best_epoch": int(best_epoch),
        "best_student_state_dict": best_student_state_dict,
        "best_teacher_state_dict": best_teacher_state_dict,
        "early_stopping_counter": int(early_stopping_counter),
        "history": history,
        "config": {
            key: value for key, value in config.items() if not str(key).startswith("_")
        },
        "mean_teacher_config": copy.deepcopy(mean_teacher_config),
    }

    torch.save(payload, path)
    print(f"[Checkpoint] Saved: {path}")
    return path


def load_mean_teacher_checkpoint(
    checkpoint_path,
    student,
    teacher,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    scheduler,
    device,
):
    checkpoint = torch.load(
        checkpoint_path,
        map_location=device,
        weights_only=False,
    )

    if checkpoint.get("training_method") != EXPERIMENT_NAME:
        raise ValueError(
            "Checkpoint belongs to a different training method."
        )

    for key, component in (
        ("student", student),
        ("teacher", teacher),
        ("supervised_criterion", supervised_criterion),
        ("consistency_criterion", consistency_criterion),
        ("optimizer", optimizer),
    ):
        validate_resume_component(checkpoint, key, component)

    if scheduler is not None:
        validate_resume_component(checkpoint, "scheduler", scheduler)

    if checkpoint["mean_teacher_config"]["contrastive"] != student.contrastive_config:
        raise ValueError("Checkpoint contrastive settings differ from the current model")
    student.load_state_dict(checkpoint["student_state_dict"])
    teacher.load_state_dict(checkpoint["teacher_state_dict"])
    supervised_criterion.load_state_dict(checkpoint["supervised_criterion_state_dict"])
    consistency_criterion.load_state_dict(
        checkpoint["consistency_criterion_state_dict"]
    )
    optimizer.load_state_dict(checkpoint["optimizer_state_dict"])

    if scheduler is not None and checkpoint.get("scheduler_state_dict") is not None:
        scheduler.load_state_dict(checkpoint["scheduler_state_dict"])

    teacher.requires_grad_(False)
    teacher.train()
    return checkpoint


def save_best_mean_teacher(
    student,
    teacher,
    best_student_state_dict,
    best_teacher_state_dict,
    best_loss,
    best_dice,
    best_epoch,
    history,
    config,
    mean_teacher_config,
):
    model_dir = Path(config["model_dir"])
    model_dir.mkdir(parents=True, exist_ok=True)
    path = (
        model_dir
        / f"best_{EXPERIMENT_NAME}_{student.__class__.__name__}_epoch_{best_epoch}.pth"
    )

    torch.save(
        {
            "training_method": EXPERIMENT_NAME,
            "student_type": component_type(student),
            "student_state_dict": best_student_state_dict,
            "teacher_type": component_type(teacher),
            "teacher_state_dict": best_teacher_state_dict,
            "best_loss": float(best_loss),
            "best_dice": float(best_dice),
            "best_epoch": int(best_epoch),
            "history": history,
            "config": {
                key: value
                for key, value in config.items()
                if not str(key).startswith("_")
            },
            "mean_teacher_config": copy.deepcopy(mean_teacher_config),
            "class_names": CLASS_NAMES,
            "class_colors": CLASS_COLORS,
        },
        path,
    )
    print(f"[Best model] Saved: {path}")
    return path
```

## One epoch


### Metric alignment with `unet.ipynb`

Training metrics are now reported with the same supervised-epoch semantics as the U-Net baseline: each labeled sample contributes exactly once per epoch, Dice is computed from epoch-level TP/FP/FN counts, and supervised loss is weighted by labeled batch size. The unlabeled loader is cycled independently only for Mean Teacher consistency training.


```python
def train_mean_teacher_one_epoch(
    student,
    teacher,
    labeled_loader,
    unlabeled_loader,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    device,
    epoch,
    global_step,
    num_classes,
    mean_teacher_config,
):
    """
    Mean Teacher epoch with U-Net-compatible metric aggregation.

    Important metric rule:
      - One epoch is defined by ONE full pass over labeled_loader.
      - Each labeled sample contributes exactly once to train Dice.
      - Supervised loss is averaged by the number of labeled samples,
        matching the supervised U-Net notebook.
      - unlabeled_loader is cycled only as needed for consistency training.

    This keeps the semi-supervised training logic while making reported
    supervised metrics directly comparable with unet.ipynb.
    """
    student.train()
    teacher.train()

    consistency_weight = get_consistency_weight(epoch, mean_teacher_config)
    ema_decay = float(mean_teacher_config["ema_decay"])

    # Match supervised U-Net epoch semantics:
    # each labeled batch is consumed exactly once.
    steps = len(labeled_loader)
    unlabeled_iterator = iter(unlabeled_loader)

    supervised_sum = 0.0
    supervised_samples = 0

    consistency_sum = 0.0
    weighted_consistency_sum = 0.0
    consistency_ratio_sum = 0.0
    total_sum = 0.0
    contrastive_sums = {name: 0.0 for name in CONTRASTIVE_METRIC_NAMES}

    epoch_tp = np.zeros(num_classes, dtype=np.float64)
    epoch_fp = np.zeros(num_classes, dtype=np.float64)
    epoch_fn = np.zeros(num_classes, dtype=np.float64)

    for labeled_batch in labeled_loader:
        unlabeled_batch, unlabeled_iterator = _next_cycled_batch(
            unlabeled_loader,
            unlabeled_iterator,
        )

        labeled_weak = labeled_batch["weak_image"].to(
            device,
            non_blocking=True,
        )
        labeled_strong = labeled_batch["strong_image"].to(
            device,
            non_blocking=True,
        )
        labeled_masks = (
            labeled_batch["mask"]
            .to(
                device,
                non_blocking=True,
            )
            .long()
        )

        unlabeled_weak = unlabeled_batch["weak_image"].to(
            device,
            non_blocking=True,
        )
        unlabeled_strong = unlabeled_batch["strong_image"].to(
            device,
            non_blocking=True,
        )

        num_labeled = labeled_strong.size(0)

        student_input = torch.cat(
            [labeled_strong, unlabeled_strong],
            dim=0,
        )
        teacher_input = torch.cat(
            [labeled_weak, unlabeled_weak],
            dim=0,
        )

        optimizer.zero_grad(set_to_none=True)

        student_logits = student(student_input)

        with torch.no_grad():
            teacher_logits = teacher(teacher_input)

        supervised_loss = supervised_criterion(
            student_logits[:num_labeled],
            labeled_masks,
        )

        consistency_loss = consistency_criterion(
            student_logits,
            teacher_logits,
        )

        weighted_consistency_loss = consistency_weight * consistency_loss

        consistency_to_supervised_ratio = (
            weighted_consistency_loss.detach()
            / supervised_loss.detach().clamp_min(1e-8)
        )

        weighted_contrastive_loss, contrastive_metrics = compute_student_contrastive(
            student, student_input, student_logits, labeled_masks,
            num_labeled, epoch, global_step,
        )
        contrastive_metrics["contrastive_to_supervised_ratio"] = (
            contrastive_metrics["weighted_contrastive_loss"]
            / max(float(supervised_loss.detach()), 1e-8)
        )
        for name, value in contrastive_metrics.items():
            contrastive_sums[name] += value
        total_loss = supervised_loss + weighted_consistency_loss + weighted_contrastive_loss

        total_loss.backward()
        torch.nn.utils.clip_grad_norm_(
            student.parameters(),
            max_norm=1.0,
        )
        optimizer.step()

        global_step += 1
        update_ema_teacher(
            student,
            teacher,
            ema_decay=ema_decay,
            global_step=global_step,
        )

        # Same Dice aggregation as unet.ipynb:
        # accumulate TP/FP/FN across the entire labeled epoch,
        # then compute foreground macro Dice once.
        with torch.no_grad():
            batch_tp, batch_fp, batch_fn = _confusion_counts_from_logits(
                student_logits[:num_labeled],
                labeled_masks,
                num_classes=num_classes,
            )
            epoch_tp += batch_tp
            epoch_fp += batch_fp
            epoch_fn += batch_fn

        # Same supervised-loss aggregation as unet.ipynb:
        # weighted by labeled batch size.
        supervised_sum += supervised_loss.item() * num_labeled
        supervised_samples += num_labeled

        # Consistency diagnostics remain step-averaged because they are
        # semi-supervised quantities over labeled + unlabeled inputs.
        consistency_sum += consistency_loss.item()
        weighted_consistency_sum += weighted_consistency_loss.item()
        consistency_ratio_sum += consistency_to_supervised_ratio.item()
        total_sum += total_loss.item()

    train_dice = _macro_dice_from_counts(
        epoch_tp,
        epoch_fp,
        epoch_fn,
        include_background=False,
    )

    metrics = {
        "supervised_loss": (supervised_sum / max(supervised_samples, 1)),
        "consistency_loss": (consistency_sum / max(steps, 1)),
        "weighted_consistency_loss": (weighted_consistency_sum / max(steps, 1)),
        "consistency_to_supervised_ratio": (consistency_ratio_sum / max(steps, 1)),
        "total_loss": (total_sum / max(steps, 1)),
        "train_dice": train_dice,
        "consistency_weight": consistency_weight,
    }

    metrics.update({name: value / max(steps, 1) for name, value in contrastive_sums.items()})
    if metrics["contrastive_weight"] > 0 and metrics["contrastive_active_fraction"] == 0:
        print("[Contrastive] No valid pairs this epoch: inspect class patch counts, "
              "pseudo coverage, patch_size and confidence/purity thresholds.")
    return metrics, global_step


@torch.no_grad()
def validate_one_epoch(
    model,
    loader,
    criterion,
    device,
    num_classes=3,
):
    model.eval()
    running_loss = 0.0
    total_samples = 0

    epoch_tp = np.zeros(num_classes, dtype=np.float64)
    epoch_fp = np.zeros(num_classes, dtype=np.float64)
    epoch_fn = np.zeros(num_classes, dtype=np.float64)

    for batch in loader:
        images = batch["image"].to(device, non_blocking=True)
        masks = batch["mask"].to(device, non_blocking=True).long()

        logits = model(images)
        loss = criterion(logits, masks)
        batch_size = images.size(0)

        batch_tp, batch_fp, batch_fn = _confusion_counts_from_logits(
            logits,
            masks,
            num_classes=num_classes,
        )
        epoch_tp += batch_tp
        epoch_fp += batch_fp
        epoch_fn += batch_fn

        running_loss += loss.item() * batch_size
        total_samples += batch_size

    epoch_dice = _macro_dice_from_counts(
        epoch_tp,
        epoch_fp,
        epoch_fn,
        include_background=False,
    )

    return (
        running_loss / max(total_samples, 1),
        epoch_dice,
    )
```

## Training loop


```python
def train_mean_teacher(
    student,
    teacher,
    labeled_loader,
    unlabeled_loader,
    val_loader,
    supervised_criterion,
    consistency_criterion,
    optimizer,
    scheduler,
    config,
    mean_teacher_config,
    checkpoint_path=None,
):
    device = torch.device(config["device"])
    target_epoch = int(config["epochs"])
    num_classes = int(config["num_classes"])
    save_every = int(config.get("save_every", 0))

    early_stopping_enabled = bool(config.get("early_stopping", False))
    patience = int(config.get("early_stopping_patience", 10))
    min_delta = float(config.get("early_stopping_min_delta", 0.0))

    if patience < 1:
        raise ValueError("early_stopping_patience must be >= 1.")
    if min_delta < 0:
        raise ValueError("early_stopping_min_delta must be >= 0.")

    history = empty_mean_teacher_history(num_classes)
    start_epoch = 1
    global_step = 0
    best_loss = float("inf")
    best_dice = float("-inf")
    best_epoch = 0
    best_student_state_dict = None
    best_teacher_state_dict = None
    early_stopping_counter = 0

    if checkpoint_path is not None:
        checkpoint = load_mean_teacher_checkpoint(
            checkpoint_path=checkpoint_path,
            student=student,
            teacher=teacher,
            supervised_criterion=supervised_criterion,
            consistency_criterion=consistency_criterion,
            optimizer=optimizer,
            scheduler=scheduler,
            device=device,
        )

        start_epoch = int(checkpoint["epoch"]) + 1
        global_step = int(checkpoint.get("global_step", 0))
        best_loss = float(checkpoint.get("best_loss", float("inf")))
        best_epoch = int(checkpoint.get("best_epoch", 0))
        best_student_state_dict = checkpoint.get("best_student_state_dict")
        best_teacher_state_dict = checkpoint.get("best_teacher_state_dict")
        early_stopping_counter = int(checkpoint.get("early_stopping_counter", 0))
        history = checkpoint.get("history", history)

        saved_best_dice = checkpoint.get("best_dice")
        if saved_best_dice is not None:
            best_dice = float(saved_best_dice)
        elif best_epoch > 0 and len(history.get("teacher_val_dice", [])) >= best_epoch:
            # Backward-compatible resume from an older checkpoint whose
            # stored best state was selected by Teacher validation loss.
            best_dice = float(history["teacher_val_dice"][best_epoch - 1])

        expected_keys = empty_mean_teacher_history(num_classes).keys()
        completed_epochs = start_epoch - 1
        for key in expected_keys:
            values = history.setdefault(key, [])
            if len(values) < completed_epochs:
                values.extend([float("nan")] * (completed_epochs - len(values)))

        print(
            f"[Resume] epoch {start_epoch - 1} -> {target_epoch} | "
            f"global_step={global_step}"
        )
    else:
        print(f"[Train] epoch 1 -> {target_epoch}")

    if start_epoch > target_epoch:
        print(
            f"[Warning] No epoch to train: start={start_epoch}, "
            f"target={target_epoch}"
        )
        # A completed training checkpoint stores both last and best weights.
        # Evaluation must use the same validation-selected weights as a normal run.
        if best_student_state_dict is not None and best_teacher_state_dict is not None:
            student.load_state_dict(best_student_state_dict)
            teacher.load_state_dict(best_teacher_state_dict)
        teacher.requires_grad_(False).eval()
        return student, teacher, history

    for epoch in range(start_epoch, target_epoch + 1):
        train_metrics, global_step = train_mean_teacher_one_epoch(
            student=student,
            teacher=teacher,
            labeled_loader=labeled_loader,
            unlabeled_loader=unlabeled_loader,
            supervised_criterion=supervised_criterion,
            consistency_criterion=consistency_criterion,
            optimizer=optimizer,
            device=device,
            epoch=epoch,
            global_step=global_step,
            num_classes=num_classes,
            mean_teacher_config=mean_teacher_config,
        )

        student_val_loss, student_val_dice = validate_one_epoch(
            student,
            val_loader,
            supervised_criterion,
            device,
            num_classes,
        )
        teacher_val_loss, teacher_val_dice = validate_one_epoch(
            teacher,
            val_loader,
            supervised_criterion,
            device,
            num_classes,
        )

        improved = teacher_val_dice > (best_dice + min_delta)
        if improved:
            best_loss = teacher_val_loss
            best_dice = teacher_val_dice
            best_epoch = epoch
            best_student_state_dict = clone_state_dict_to_cpu(student)
            best_teacher_state_dict = clone_state_dict_to_cpu(teacher)
            early_stopping_counter = 0
        elif early_stopping_enabled:
            early_stopping_counter += 1

        if scheduler is not None:
            if isinstance(scheduler, torch.optim.lr_scheduler.ReduceLROnPlateau):
                scheduler.step(teacher_val_loss)
            else:
                scheduler.step()

        current_lr = optimizer.param_groups[0]["lr"]

        for key in (
            "supervised_loss",
            "consistency_loss",
            "weighted_consistency_loss",
            "consistency_to_supervised_ratio",
            "total_loss",
            "train_dice",
            "consistency_weight",
        ):
            history[key].append(train_metrics[key])

        for name in CONTRASTIVE_METRIC_NAMES:
            history[name].append(train_metrics[name])
        history["student_val_loss"].append(student_val_loss)
        history["student_val_dice"].append(student_val_dice)
        history["teacher_val_loss"].append(teacher_val_loss)
        history["teacher_val_dice"].append(teacher_val_dice)
        history["learning_rate"].append(current_lr)

        print(
            f"Epoch {epoch:02d}/{target_epoch:02d} | "
            f"sup={train_metrics['supervised_loss']:.5f} | "
            f"cons={train_metrics['consistency_loss']:.6f} | "
            f"cl={train_metrics['contrastive_loss']:.5f} | "
            f"w_cl={train_metrics['weighted_contrastive_loss']:.5f} | "
            f"cl_active={train_metrics['contrastive_active_fraction']:.2f} | "
            f"anchors={train_metrics['contrastive_anchors']:.1f} | "
            f"w_cons={train_metrics['weighted_consistency_loss']:.6f} | "
            f"w_cons/sup={train_metrics['consistency_to_supervised_ratio']:.4f} | "
            f"lambda_c={train_metrics['consistency_weight']:.4f} | "
            f"student_val_dice={student_val_dice:.4f} | "
            f"teacher_val_dice={teacher_val_dice:.4f} | "
            f"lr={current_lr:.7f} | "
            f"best_teacher_val_dice={best_dice:.4f} @ {best_epoch}"
        )

        output_dir = Path(config["model_dir"])
        output_dir.mkdir(parents=True, exist_ok=True)
        pd.DataFrame(history).to_csv(output_dir / "history.csv", index=False)
        if ((save_every > 0 and epoch % save_every == 0) or epoch == target_epoch
                or (early_stopping_enabled and early_stopping_counter >= patience)):
            save_mean_teacher_checkpoint(
                student=student,
                teacher=teacher,
                supervised_criterion=supervised_criterion,
                consistency_criterion=consistency_criterion,
                optimizer=optimizer,
                scheduler=scheduler,
                epoch=epoch,
                global_step=global_step,
                best_loss=best_loss,
                best_dice=best_dice,
                best_epoch=best_epoch,
                best_student_state_dict=best_student_state_dict,
                best_teacher_state_dict=best_teacher_state_dict,
                early_stopping_counter=early_stopping_counter,
                history=history,
                config=config,
                mean_teacher_config=mean_teacher_config,
                checkpoint_dir=config["checkpoint_dir"],
            )

        if early_stopping_enabled and early_stopping_counter >= patience:
            print(
                f"[Early stopping] No Teacher validation-Dice improvement "
                f"for {patience} epochs. Best epoch: {best_epoch}."
            )
            break

    if best_teacher_state_dict is None or best_student_state_dict is None:
        best_student_state_dict = clone_state_dict_to_cpu(student)
        best_teacher_state_dict = clone_state_dict_to_cpu(teacher)
        best_epoch = epoch
        best_loss = teacher_val_loss
        best_dice = teacher_val_dice

    save_best_mean_teacher(
        student=student,
        teacher=teacher,
        best_student_state_dict=best_student_state_dict,
        best_teacher_state_dict=best_teacher_state_dict,
        best_loss=best_loss,
        best_dice=best_dice,
        best_epoch=best_epoch,
        history=history,
        config=config,
        mean_teacher_config=mean_teacher_config,
    )

    student.load_state_dict(best_student_state_dict)
    teacher.load_state_dict(best_teacher_state_dict)
    student.to(device)
    teacher.to(device).requires_grad_(False).eval()
    return student, teacher, history
```

```python
def plot_mean_teacher_history(history):
    epochs = np.arange(1, len(history["total_loss"]) + 1)
    fig, axes = plt.subplots(2, 2, figsize=(16, 11))

    axes[0, 0].plot(epochs, history["supervised_loss"], label="Supervised")
    axes[0, 0].plot(epochs, history["consistency_loss"], label="Consistency (raw)")
    axes[0, 0].plot(
        epochs, history["weighted_consistency_loss"], label="Consistency (weighted)"
    )
    axes[0, 0].plot(epochs, history["total_loss"], label="Total")
    axes[0, 0].set_title("Training losses")
    axes[0, 0].legend()

    axes[0, 1].plot(epochs, history["train_dice"], label="Student train")
    axes[0, 1].plot(epochs, history["student_val_dice"], label="Student validation")
    axes[0, 1].plot(epochs, history["teacher_val_dice"], label="Teacher validation")
    axes[0, 1].set_title("Foreground Dice")
    axes[0, 1].legend()

    axes[1, 0].plot(
        epochs,
        history["student_val_loss"],
        label="Student validation",
    )
    axes[1, 0].plot(
        epochs,
        history["teacher_val_loss"],
        label="Teacher validation",
    )
    axes[1, 0].set_title("Validation supervised loss")
    axes[1, 0].legend()

    axes[1, 1].plot(
        epochs,
        history["consistency_weight"],
        label="Consistency weight",
    )
    axes[1, 1].plot(
        epochs,
        history["consistency_to_supervised_ratio"],
        label="Weighted consistency / supervised",
    )
    ax_lr = axes[1, 1].twinx()
    ax_lr.plot(
        epochs,
        history["learning_rate"],
        color="tab:orange",
        label="Learning rate",
    )
    axes[1, 1].set_title("Consistency weight and learning rate")
    axes[1, 1].set_ylabel("Consistency diagnostics")
    axes[1, 1].legend(loc="upper left")
    ax_lr.set_ylabel("Learning rate")

    for axis in axes.flat:
        axis.set_xlabel("Epoch")
        axis.grid(alpha=0.3)

    plt.tight_layout()
    plt.show()

def plot_contrastive_history(history):
    epochs = np.arange(1, len(history["total_loss"]) + 1)
    fig, axes = plt.subplots(2, 2, figsize=(14, 9))
    for key in ("contrastive_loss", "weighted_contrastive_loss"):
        axes[0, 0].plot(epochs, history[key], label=key)
    for key in ("contrastive_weight", "contrastive_to_supervised_ratio"):
        axes[0, 1].plot(epochs, history[key], label=key)
    for key in ("pseudo_coverage", "contrastive_active_fraction"):
        axes[1, 0].plot(epochs, history[key], label=key)
    for key in ("patches_background", "patches_non_neoplastic", "patches_neoplastic"):
        axes[1, 1].plot(epochs, history[key], label=key)
    for ax in axes.flat:
        ax.set_xlabel("Epoch")
        ax.legend()
        ax.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
@torch.no_grad()
def evaluate_model(
    model,
    loader,
    device,
    num_classes=3,
    class_names=None,
):
    """
    Evaluate model on a dataset and compute metrics per class.

    Args:
        model: model to evaluate
        loader: DataLoader of the evaluation dataset
        device: device used for evaluation
        num_classes: number of segmentation classes
        class_names: dictionary mapping class IDs to class names

    Returns:
        DataFrame containing:
            - metrics for each class
            - macro_foreground_only:
                average of foreground classes only
            - macro_all_classes:
                average of all classes including background
    """

    class_names = class_names or {i: f"class_{i}" for i in range(num_classes)}

    tp = np.zeros(
        num_classes,
        dtype=np.float64,
    )

    fp = np.zeros(
        num_classes,
        dtype=np.float64,
    )

    fn = np.zeros(
        num_classes,
        dtype=np.float64,
    )

    hd95_values = {class_id: [] for class_id in range(num_classes)}

    model.eval()

    for batch in loader:
        images = batch["image"].to(
            device,
            non_blocking=True,
        )

        masks = (
            batch["mask"]
            .to(
                device,
                non_blocking=True,
            )
            .long()
        )

        logits = model(images)

        preds = torch.argmax(
            logits,
            dim=1,
        )

        preds_np = preds.cpu().numpy()
        masks_np = masks.cpu().numpy()

        for class_id in range(num_classes):
            pred_c = preds_np == class_id
            true_c = masks_np == class_id

            tp[class_id] += np.logical_and(
                pred_c,
                true_c,
            ).sum()

            fp[class_id] += np.logical_and(
                pred_c,
                ~true_c,
            ).sum()

            fn[class_id] += np.logical_and(
                ~pred_c,
                true_c,
            ).sum()

            for sample_idx in range(preds_np.shape[0]):
                hd95_values[class_id].append(
                    hd95_binary(
                        pred_c[sample_idx],
                        true_c[sample_idx],
                    )
                )

    rows = []

    for class_id in range(num_classes):
        dice = _safe_divide(
            2.0 * tp[class_id],
            (2.0 * tp[class_id] + fp[class_id] + fn[class_id]),
        )

        iou = _safe_divide(
            tp[class_id],
            (tp[class_id] + fp[class_id] + fn[class_id]),
        )

        precision = _safe_divide(
            tp[class_id],
            (tp[class_id] + fp[class_id]),
        )

        recall = _safe_divide(
            tp[class_id],
            (tp[class_id] + fn[class_id]),
        )

        hd_values = np.asarray(
            hd95_values[class_id],
            dtype=np.float64,
        )
        valid_hd_values = hd_values[~np.isnan(hd_values)]

        hd95 = (
            float(valid_hd_values.mean())
            if valid_hd_values.size > 0
            else np.nan
        )

        rows.append(
            {
                "class_id": class_id,
                "class_name": class_names.get(
                    class_id,
                    f"class_{class_id}",
                ),
                "dice": dice,
                "iou": iou,
                "hd95": hd95,
                "precision": precision,
                "recall": recall,
            }
        )

    result = pd.DataFrame(rows)

    # Macro foreground only
    foreground = result[result["class_id"].isin(range(1, num_classes))]

    macro_fg_row = {
        "class_id": "macro_fg",
        "class_name": "macro_foreground_only",
        "dice": foreground["dice"].mean(skipna=True),
        "iou": foreground["iou"].mean(skipna=True),
        "hd95": foreground["hd95"].mean(skipna=True),
        "precision": foreground["precision"].mean(skipna=True),
        "recall": foreground["recall"].mean(skipna=True),
    }

    # Macro all classes includes background
    all_classes = result[result["class_id"].isin(range(num_classes))]

    macro_all_row = {
        "class_id": "macro_all",
        "class_name": "macro_all_classes",
        "dice": all_classes["dice"].mean(skipna=True),
        "iou": all_classes["iou"].mean(skipna=True),
        "hd95": all_classes["hd95"].mean(skipna=True),
        "precision": all_classes["precision"].mean(skipna=True),
        "recall": all_classes["recall"].mean(skipna=True),
    }

    return pd.concat(
        [
            result,
            pd.DataFrame(
                [
                    macro_fg_row,
                    macro_all_row,
                ]
            ),
        ],
        ignore_index=True,
    )
```

```python
# Run after BOTH experiments finish; no training or test-based checkpoint selection here.
def compare_contrastive_runs(models_root, run_id):
    runs = {}
    for objective in ("infonce", "plgcl"):
        folder = Path(models_root) / f"mean_teacher_student_{objective}" / run_id
        required = ("experiment.json", "teacher_test_metrics.csv", "history.csv")
        if not all((folder / name).is_file() for name in required):
            print(f"Comparison pending: finish {objective} for {run_id} first.")
            return None
        runs[objective] = (
            folder,
            json.loads((folder / "experiment.json").read_text(encoding="utf-8")),
        )
    a, b = (runs[key][1] for key in ("infonce", "plgcl"))
    # Methods/paths differ by design. Other training settings must match.
    ignored = {
        "checkpoint_dir",
        "model_dir",
        "device",
        "save_every",
        "timestamp_format",
    }
    comparable_config = lambda m: {
        k: v for k, v in m["config"].items() if k not in ignored
    }
    comparable_mt = lambda m: {
        **m["mean_teacher_config"],
        "contrastive": {
            k: v
            for k, v in m["mean_teacher_config"]["contrastive"].items()
            if k != "objective"
        },
    }
    if (
        a["split_files_sha256"] != b["split_files_sha256"]
        or comparable_config(a) != comparable_config(b)
        or comparable_mt(a) != comparable_mt(b)
        or a["selection_metric"] != b["selection_metric"]
    ):
        raise ValueError(
            "Experiments have different splits/configuration; do not treat this as a loss-only comparison."
        )
    frames = []
    for objective, (folder, manifest) in runs.items():
        history_frame = pd.read_csv(folder / "history.csv")
        if len(history_frame) != int(manifest["config"]["epochs"]):
            raise ValueError(
                f"{objective} did not complete its configured training budget."
            )
        metrics = pd.read_csv(folder / "teacher_test_metrics.csv")
        metrics.insert(0, "method", objective)
        frames.append(metrics)
    comparison = pd.concat(frames, ignore_index=True)
    display(comparison[["method", "class_name", "dice", "iou", "hd95"]].round(4))
    print(
        "Higher Dice/IoU and lower HD95 are better. Report several seeds before drawing conclusions."
    )
    return comparison


comparison = compare_contrastive_runs(PROJECT_ROOT / "models_contrastive", run_id)
```
